<div class="card" dir="rtl">
<h2>لیست (List): اولین و مهم‌ترین!</h2>
<p>
لیست‌ها در پایتون مانند یک فهرست یا یک لیست خرید هستند. شما می‌توانید هر چیزی را به هر ترتیبی که دوست دارید داخل آن قرار دهید و بعداً ترتیب آن‌ها را عوض کنید، چیزی را حذف کنید یا یک آیتم جدید اضافه کنید.
</p>
<p>
<strong>ویژگی‌های کلیدی لیست‌ها:</strong>
</p>
<ul>
<li><strong>منظم (Ordered):</strong> ترتیب قرار گرفتن عناصر در لیست مهم است و تغییر نمی‌کند (مگر اینکه خودمان تغییر دهیم).</li>
<li><strong>قابل تغییر (Mutable):</strong> می‌توانیم عناصر لیست را بعد از ایجاد، اضافه، حذف یا ویرایش کنیم.</li>
<li><strong>اعضای تکراری مجاز است:</strong> یک لیست می‌تواند شامل چندین عضو با مقدار یکسان باشد.</li>
<li><strong>انواع داده‌های مختلف:</strong> می‌توانیم در یک لیست، انواع مختلفی از داده‌ها (عدد، رشته، و حتی لیست دیگر) را در کنار هم داشته باشیم.</li>
</ul>
<p>
یک آنالوژی خوب برای لیست‌ها، یک <strong>دستور پخت شماره‌گذاری شده</strong> است. هر مرحله از دستور پخت (عنصر لیست) به ترتیب و با شماره‌ی مشخصی (اندیس) آمده است. مهم‌تر اینکه، شما می‌توانید این دستور پخت را تغییر دهید: یک مرحله را ویرایش کنید، یک مرحله جدید بین مراحل دیگر اضافه نمایید، یا حتی یک مرحله را به کلی حذف کنید.
</p>

<div class="card" dir="rtl">
<h2>کالبدشکافی لیست‌ها: اندیس، طول و برش</h2>
<p>
لیست‌ها فقط مجموعه‌ای از آیتم‌ها نیستند؛ آن‌ها یک ساختار منظم دارند که به ما اجازه می‌دهد با دقت به هر قسمتی از آن‌ها دسترسی داشته باشیم. بیایید با سه مفهوم کلیدی در کار با لیست‌ها آشنا شویم:
</p>
<h4>۱. طول لیست (Length)</h4>
<p>
برای فهمیدن اینکه یک لیست چند عضو دارد، از تابع <code>len()</code> استفاده می‌کنیم. این تابع مانند یک متر، تعداد کل آیتم‌های داخل لیست را به ما می‌دهد.
</p>
<h4>۲. اندیس‌گذاری (Indexing)</h4>
<p>
هر عضو در یک لیست، یک آدرس یا شماره منحصر به فرد به نام <strong>اندیس (Index)</strong> دارد. این شماره‌گذاری به دو صورت انجام می‌شود:
</p>
<ul>
<li><strong>اندیس مثبت:</strong> از <strong>۰</strong> برای اولین عضو شروع می‌شود و به سمت راست افزایش می‌یابد (۰, ۱, ۲, ...).</li>
<li><strong>اندیس منفی:</strong> از <strong>۱-</strong> برای آخرین عضو شروع می‌شود و به سمت چپ حرکت می‌کند (۱-, ۲-, ۳-, ...). اندیس منفی یک راه سریع برای دسترسی به اعضای انتهای لیست است.</li>
</ul>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/list.png" alt="Using Variables" style="max-height: 250px;">
</div>
<h4>۳. برش زدن (Slicing)</h4>
<p>
اسلایسینگ به ما این امکان را می‌دهد که یک <strong>زیرمجموعه یا برشی</strong> از لیست اصلی را استخراج کنیم. این کار با فرمت <code>[start:stop:step]</code> انجام می‌شود.
</p>
<ul>
<li><code>start</code>: اندیس شروع برش (شامل خود این اندیس می‌شود).</li>
<li><code>stop</code>: اندیس پایان برش (شامل این اندیس <strong>نمی‌شود</strong>).</li>
<li><code>step</code>: گام حرکت (مثلاً ۲ یعنی یکی در میان).</li>
</ul>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/list_slicing.png" alt="Using Variables" style="max-height: 250px;">
</div>
</div>


```python
letters = ['a', 'b', 'c', 'd', 'e', 'f']
# 1. طول لیست
print(len(letters)) # خروجی: 6

# 2. اندیس‌گذاری
print(letters[0]) # اولین عضو: 'a'
print(letters[2]) # سومین عضو: 'c'
print(letters[-1]) # آخرین عضو: 'f'
print(letters[-2]) # یکی مانده به آخر: 'e'

# 3. برش زدن (Slicing)
# از اندیس 1 تا قبل از اندیس 4
print(letters[1:4]) # خروجی: ['b', 'c', 'd']
# از ابتدا تا قبل از اندیس 3
print(letters[:3]) # خروجی: ['a', 'b', 'c']
# از اندیس 2 تا انتها
print(letters[2:]) # خروجی: ['c', 'd', 'e', 'f']
# برعکس کردن لیست با گام -1
print(letters[::-1]) # خروجی: ['f', 'e', 'd', 'c', 'b', 'a']
```

<!-- اسلاید جدید: ساخت و کار با لیست‌ها -->
<div class="card" dir="rtl">
<h2>ایجاد لیست و آشنایی با قابلیت‌های آن</h2>
<h4>چگونه یک لیست بسازیم؟</h4>
<p>
ساده‌ترین راه برای ساختن یک لیست در پایتون، استفاده از براکت <code>[]</code> است. شما می‌توانید آیتم‌های خود را داخل براکت قرار داده و آن‌ها را با کاما (<code>,</code>) از یکدیگر جدا کنید. یک لیست می‌تواند خالی باشد، یا شامل انواع مختلفی از داده‌ها مانند اعداد، رشته‌ها و حتی لیست‌های دیگر باشد. لازم به ذکر است که شما می‌توانید آیتم‌هایی با دیتاتایپ مختلف (مثلاً عدد صحیح، عدد اعشاری، رشته یا ...) را در یک لیست داشته باشید. در واقع یک لیست محدود به نگهداری آیتم‌هایی از یک دیتاتایپ یکسان نیست.
</p>
<h4>توابع و متدهای کلیدی لیست‌ها (مرور کامل)</h4>
<p>در اینجا مجموعه‌ای کامل از ابزارهایی که برای کار با لیست‌ها در اختیار دارید، آمده است:</p>
<ul>
<li><b>برای اضافه کردن:</b>
<ul>
<li><code>append(x)</code>: آیتم <code>x</code> را به انتهای لیست اضافه می‌کند.</li>
<li><code>insert(i, x)</code>: آیتم <code>x</code> را در موقعیت <code>i</code> قرار می‌دهد.</li>
<li><code>extend(iterable)</code>: تمام اعضای یک مجموعه دیگر (مانند یک لیست دیگر) را به انتهای لیست فعلی اضافه می‌کند.</li>
</ul>
</li>
<li><b>برای حذف کردن:</b>
<ul>
<li><code>remove(x)</code>: اولین آیتمی که مقدارش <code>x</code> باشد را حذف می‌کند.</li>
<li><code>pop(i)</code>: آیتم موقعیت <code>i</code> را حذف و آن را برمی‌گرداند (اگر <code>i</code> داده نشود، آخرین آیتم را حذف می‌کند).</li>
<li><code>clear()</code>: تمام آیتم‌های لیست را پاک کرده و آن را خالی می‌کند.</li>
<li><code>del my_list[i]</code>: آیتم در موقعیت <code>i</code> را حذف می‌کند (این یک کلمه کلیدی است نه متد).</li>
</ul>
</li>
<li><b>برای جستجو و شمارش:</b>
<ul>
<li><code>index(x)</code>: شماره اندیس اولین آیتمی که مقدارش <code>x</code> است را برمی‌گرداند.</li>
<li><code>count(x)</code>: تعداد دفعاتی که آیتم <code>x</code> در لیست تکرار شده است را می‌شمارد.</li>
</ul>
</li>
<li><b>برای مرتب‌سازی و سازماندهی:</b>
<ul>
<li><code>sort()</code>: لیست را <strong>درجا</strong> مرتب می‌کند (از کوچک به بزرگ).</li>
<li><code>reverse()</code>: ترتیب اعضای لیست را <strong>درجا</strong> معکوس می‌کند.</li>
<li><code>copy()</code>: یک کپی از لیست برمی‌گرداند تا لیست اصلی دست‌نخورده باقی بماند.</li>
</ul>
</li>
</ul>
</div>
</div>

```python
# 1. ساخت یک لیست
numbers = [10, 50, 70]
print(f"لیست اولیه: {numbers}")

# 2. اضافه کردن آیتم‌ها
numbers.append(20)         # -> [10, 50, 70, 20]
numbers.insert(1, 15)      # -> [10, 15, 50, 70, 20]
numbers.extend([30, 40])   # -> [10, 15, 50, 70, 20, 30, 40]
print(f"بعد از اضافه کردن: {numbers}")

# 3. حذف آیتم‌ها
numbers.remove(50)               # -> [10, 15, 70, 20, 30, 40]
popped_item = numbers.pop()      # popped_item = 40
print(f"بعد از حذف: {numbers}")
print(f"آیتم حذف شده با pop: {popped_item}")

# 4. جستجو و شمارش
print(f"اندیس عدد 10: {numbers.index(10)}")   # -> 0
numbers.append(20)
print(f"تعداد تکرار عدد 20: {numbers.count(20)}")  # -> 2

# 5. مرتب‌سازی و سازماندهی
numbers.sort()          # لیست درجا مرتب می‌شود
print(f"لیست مرتب‌شده: {numbers}")
numbers.reverse()       # لیست درجا معکوس می‌شود
print(f"لیست معکوس‌شده: {numbers}")

# 6. کپی کردن
numbers_copy = numbers.copy()
numbers_copy.clear()
print(f"کپی که خالی شده: {numbers_copy}")     # -> []
print(f"لیست اصلی بدون تغییر: {numbers}")
```

<div class="card" dir="rtl">
<h2>تاپل (Tuple): برادرِ لیست</h2>
<h4>تاپل چیست و چگونه ساخته می‌شود؟</h4>
<p>
تاپل یک ساختمان داده است که بسیار شبیه به لیست عمل می‌کند، اما با یک قانون بزرگ و مهم: <strong>تاپل‌ها غیرقابل تغییر (Immutable) هستند</strong>. پس از اینکه یک تاپل را ساختید، دیگر نمی‌توانید هیچ‌کدام از اعضای آن را تغییر دهید، حذف کنید یا عضو جدیدی به آن اضافه کنید. این ویژگی، تاپل‌ها را برای نگهداری داده‌هایی که می‌خواهید ثابت و دست‌نخورده باقی بمانند، ایده‌آل می‌سازد.
</p>
<p>
<strong>ویژگی‌های کلیدی تاپل‌ها:</strong>
</p>
<ul>
<li><strong>منظم (Ordered):</strong> ترتیب عناصر در تاپل حفظ می‌شود.</li>
<li><strong>غیرقابل تغییر (Immutable):</strong> پس از ایجاد، نمی‌توان هیچ عضوی را اضافه، حذف یا ویرایش کرد.</li>
<li><strong>سریع‌تر از لیست‌ها:</strong> به دلیل غیرقابل تغییر بودن، پردازش تاپل‌ها معمولاً سریع‌تر از لیست‌ها است.</li>
<li><strong>اعضای تکراری مجاز است:</strong> درست مانند لیست‌ها.</li>
</ul>

<!-- اسلاید جدید: همه چیز درباره تاپل‌ها -->
<div class="card" dir="rtl">
<h2>ساخت تاپل‌ها</h2>

<p>تاپل‌ها معمولاً با پرانتز <code>()</code> ساخته می‌شوند. البته پرانتز همیشه اجباری نیست و پایتون می‌تواند از روی کاماها تشخیص دهد که شما در حال ساخت یک تاپل هستید (این کار را Tuple Packing می‌نامند).</p>
<h4>ویژگی‌ها و متدهای کاربردی تاپل</h4>
<p>از آنجایی که تاپل‌ها غیرقابل تغییرند، متدهایی برای اضافه یا حذف کردن آیتم ندارند. ابزارهای اصلی ما برای کار با آن‌ها، جستجو و دسترسی به اطلاعات است.</p>
<ul>
<li><b>دسترسی به اعضا:</b>
<ul>
<li><b>اندیس‌گذاری و اسلایسینگ:</b> دقیقاً مانند لیست‌ها، می‌توانید با اندیس‌های مثبت و منفی و همچنین با برش زدن (Slicing) به اعضای تاپل دسترسی داشته باشید.</li>
</ul>
</li>
<li><b>متدهای اصلی تاپل:</b>
<ul>
<li><code>count(x)</code>: تعداد دفعاتی که آیتم <code>x</code> در تاپل تکرار شده است را برمی‌گرداند.</li>
<li><code>index(x)</code>: شماره اندیس اولین عضوی که مقدارش <code>x</code> است را پیدا می‌کند.</li>
</ul>
</li>
<li><b>توابع عمومی کاربردی:</b>
<ul>
<li><code>len(my_tuple)</code>: تعداد اعضای تاپل را برمی‌گرداند.</li>
<li><code>sorted(my_tuple)</code>: یک <strong>لیست جدید مرتب‌شده</strong> از اعضای تاپل برمی‌گرداند (تاپل اصلی بدون تغییر باقی می‌ماند).</li>
</ul>
</li>
<li><b>باز کردن تاپل (Tuple Unpacking):</b>
<ul>
<li>این یکی از قابلیت‌های قدرتمند پایتون است. شما می‌توانید اعضای یک تاپل را به صورت مستقیم به چندین متغیر اختصاص دهید.</li>
</ul>
</li>
</ul>
<h4>چرا از تاپل استفاده کنیم؟</h4>
<ul>
<li><b>امنیت داده:</b> وقتی نمی‌خواهید داده‌ها به اشتباه تغییر کنند (مانند تنظیمات یک برنامه).</li>
<li><b>استفاده به عنوان کلید دیکشنری:</b> چون غیرقابل تغییر هستند، می‌توان از تاپل‌ها به عنوان کلید در دیکشنری استفاده کرد، اما از لیست‌ها نمی‌توان.</li>
<li><b>بهینه‌سازی جزئی:</b> پردازش تاپل‌ها می‌تواند اندکی سریع‌تر از لیست‌ها باشد.</li>
</ul>

```python
# ساخت یک تاپل
coordinates = (35.6892, 51.3890)
print(coordinates)

# دسترسی به یک عضو
print(coordinates[0]) # خروجی: 35.6892
print(coordinates[1]) # خروجی: 51.3890

# تلاش برای تغییر یک عضو باعث خطا می‌شود!
coordinates[0] = 40.7128 # TypeError: 'tuple' object does not support item assignment
```

<div class="card" dir="rtl">
<h2>مجموعه (Set): کیسه‌ای از آیتم‌های منحصربه‌فرد</h2>
<p>
مجموعه‌ها یا `Set` در پایتون، مانند یک کیسه هستند که شما آیتم‌ها را داخل آن می‌ریزید. ترتیب آیتم‌ها در این کیسه مهم نیست و شما نمی‌توانید دو آیتم یکسان را در آن داشته باشید. اگر سعی کنید یک آیتم تکراری اضافه کنید، نادیده گرفته می‌شود.
</p>
<p>
این ویژگی، مجموعه‌ها را برای کارهایی مثل حذف موارد تکراری از یک لیست یا بررسی عضویت یک آیتم، بسیار کارآمد می‌کند.
</p>
<p>
<strong>ویژگی‌های کلیدی مجموعه‌ها:</strong>
</p>
<ul>
<li><strong>نامنظم (Unordered):</strong> ترتیب خاصی برای عناصر وجود ندارد.</li>
<li><strong>بدون عضو تکراری (Unique):</strong> هر عنصر فقط یک بار می‌تواند در مجموعه وجود داشته باشد.</li>
<li><strong>قابل تغییر (Mutable):</strong> می‌توان آیتم‌هایی را به مجموعه اضافه یا از آن حذف کرد (اما نمی‌توان آیتم‌های موجود را ویرایش کرد).</li>
<li><strong>عملیات ریاضی مجموعه‌ها:</strong> از عملیاتی مانند اجتماع، اشتراک و تفاضل پشتیبانی می‌کنند.</li>
</ul>

<!-- اسلاید 8: مجموعه (Set) - بخش دوم: مثال -->
<div class="card" dir="rtl">
  <h2>کار با مجموعه‌ها در پایتون</h2>
  <p>
    مجموعه‌ها با استفاده از آکولاد <code>{}</code> یا تابع <code>set()</code> ساخته می‌شوند.
    ویژگی مهم مجموعه‌ها این است که <b>اعضای تکراری را حذف می‌کنند</b> و ترتیب خاصی ندارند.
  </p>

  <h4>توابع و عملیات پرکاربرد روی مجموعه‌ها:</h4>
  <ul>
    <li><code>add(x)</code>: اضافه کردن عضو جدید به مجموعه</li>
    <li><code>remove(x)</code>: حذف عضو (در صورت نبودن خطا می‌دهد)</li>
    <li><code>discard(x)</code>: حذف عضو (در صورت نبودن خطا نمی‌دهد)</li>
    <li><code>pop()</code>: حذف و بازگرداندن یک عضو تصادفی</li>
    <li><code>clear()</code>: پاک کردن کل مجموعه</li>
    <li><code>len(s)</code>: تعداد اعضای مجموعه</li>
    <li><code>x in s</code>: بررسی وجود عضو در مجموعه</li>
  </ul>

  <h4>عملیات بین مجموعه‌ها:</h4>
  <ul>
    <li><code>A.union(B)</code> یا <code>A | B</code>: اجتماع</li>
    <li><code>A.intersection(B)</code> یا <code>A & B</code>: اشتراک</li>
    <li><code>A.difference(B)</code> یا <code>A - B</code>: تفاضل</li>
    <li><code>A.symmetric_difference(B)</code> یا <code>A ^ B</code>: تفاضل متقارن</li>
    <li><code>A.issubset(B)</code>: بررسی زیرمجموعه بودن</li>
    <li><code>A.issuperset(B)</code>: بررسی ابرمجموعه بودن</li>
    <li><code>A.isdisjoint(B)</code>: بررسی نداشتن اشتراک</li>
  </ul>

  <p>
    مثال:  
    <code>{1, 2, 3} | {3, 4}</code> → <code>{1, 2, 3, 4}</code>
  </p>
</div>

```python
# ساخت یک مجموعه از یک لیست با اعضای تکراری
numbers = [1, 2, 2, 3, 4, 4, 4]
unique_numbers = set(numbers)
print("Unique numbers:", unique_numbers)  # {1, 2, 3, 4}

# اضافه کردن یک عضو
unique_numbers.add(5)
print("After adding 5:", unique_numbers)

# اضافه کردن عضو تکراری (نادیده گرفته می‌شود)
unique_numbers.add(2)
print("After adding duplicate 2:", unique_numbers)

# حذف یک عضو
unique_numbers.remove(3)  
print("After removing 3:", unique_numbers)

# حذف بدون خطا (اگر وجود نداشت خطا نمی‌دهد)
unique_numbers.discard(100)
print("After discarding 100 (not exist):", unique_numbers)

# گرفتن و حذف یک عضو تصادفی
popped = unique_numbers.pop()
print("Popped element:", popped)
print("After pop:", unique_numbers)

# بررسی وجود یک عضو
print("Is 4 in set?", 4 in unique_numbers)

# اندازه مجموعه
print("Set size:", len(unique_numbers))

# پاک کردن کل مجموعه
unique_numbers.clear()
print("After clear:", unique_numbers)  # {}
```

```python
# ساخت یک مجموعه دستی
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("A:", A)
print("B:", B)

# اجتماع (union)
print("A ∪ B:", A.union(B))

# اشتراک (intersection)
print("A ∩ B:", A.intersection(B))

# تفاضل (difference)
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

# تفاضل متقارن (symmetric difference)
print("A △ B:", A.symmetric_difference(B))

# بررسی زیرمجموعه و ابرمجموعه
print("A ⊆ B ?", A.issubset(B))
print("A ⊇ B ?", A.issuperset(B))

# بررسی نداشتن اشتراک
print("A ∩ B == ∅ ?", A.isdisjoint(B))
```

<!-- اسلاید 10: دیکشنری (Dictionary) - بخش دوم: مثال -->
<div class="card" dir="rtl">
<h2>دیکشنری در عمل</h2>
<p>
دیکشنری‌ها با آکولاد <code>{}</code> ساخته می‌شوند و هر عضو شامل <b>کلید و مقدار</b> است: <code>{key: value}</code>.
ویژگی مهم دیکشنری‌ها این است که <b>کلیدها یکتا هستند</b> و ترتیب خاصی ندارند.
</p>
<h4>متدها و قابلیت‌های مفید دیکشنری</h4>
<ul>
    <li><code>keys()</code>: دریافت لیست کلیدها</li>
    <li><code>values()</code>: دریافت لیست مقادیر</li>
    <li><code>items()</code>: دریافت لیست زوج‌های (کلید, مقدار)</li>
    <li><code>get(key, default)</code>: دسترسی امن به مقدار، بدون خطا در صورت نبود کلید</li>
    <li><code>pop(key)</code>: حذف و برگرداندن مقدار کلید</li>
    <li><code>update(other_dict)</code>: اضافه یا به‌روزرسانی کلیدها و مقادیر از یک دیکشنری دیگر</li>
    <li><code>clear()</code>: پاک کردن تمام کلیدها و مقادیر</li>
</ul>
</div>

```python
print("--- ۱. روش‌های مختلف ساخت دیکشنری ---")
# روش اول: استفاده از آکولاد {} (رایج‌ترین)
user = {"name": "Sara", "age": 25, "city": "Shiraz"}
print(f"ساخت با آکولاد: {user}")

# روش دوم: استفاده از تابع dict() با آرگومان‌های کلیدواژه‌ای
user_alt1 = dict(name="Sara", age=25, city="Shiraz")
print(f"ساخت با تابع dict: {user_alt1}")

# روش سوم: استفاده از تابع dict() با لیستی از تاپل‌های کلید-مقدار
user_alt2 = dict([('name', 'Sara'), ('age', 25), ('city', 'Shiraz')])
print(f"ساخت با لیست تاپل‌ها: {user_alt2}")
```

```python
print("--- ۲. دسترسی به مقادیر: تفاوت [] و .get() ---")
# دسترسی مستقیم با براکت: سریع است اما اگر کلید وجود نداشته باشد، KeyError می‌دهد.
print(f"دسترسی به نام: {user['name']}")
try:
    print(user['job'])
except KeyError as e:
    print(f"خطای مورد انتظار هنگام دسترسی با براکت: کلید {e} وجود ندارد.")

# دسترسی امن با متد .get(): اگر کلید وجود نداشته باشد، None یا یک مقدار پیش‌فرض برمی‌گرداند.
job = user.get('job')
print(f"دسترسی به شغل با .get(): {job}") # خروجی None است چون وجود ندارد
job_default = user.get('job', 'Unemployed')
print(f"دسترسی به شغل با مقدار پیش‌فرض: {job_default}")
```

```python
print("--- ۳. افزودن، تغییر و به‌روزرسانی ---")
print(f"دیکشنری اولیه: {user}")
# تغییر یک مقدار موجود
user['age'] = 26
print(f"بعد از تغییر سن: {user}")

# افزودن یک کلید-مقدار جدید
user['email'] = "sara@example.com"
print(f"بعد از افزودن ایمیل: {user}")

# به‌روزرسانی با یک دیکشنری دیگر با متد .update()
# این متد کلیدهای موجود را به‌روز می‌کند و کلیدهای جدید را اضافه می‌کند.
new_data = {'city': 'Isfahan', 'status': 'active'}
user.update(new_data)
print(f"بعد از به‌روزرسانی: {user}")
```

```python
print("--- ۴. حذف اعضا ---")
# حذف با del: ساده و مستقیم
del user['city']
print(f"بعد از حذف شهر با del: {user}")

# حذف با .pop(): کلید را حذف کرده و مقدار آن را برمی‌گرداند (برای استفاده بعدی)
removed_age = user.pop('age')
print(f"سن حذف شده با .pop(): {removed_age}")
print(f"دیکشنری بعد از .pop('age'): {user}")

# .popitem(): آخرین زوج کلید-مقدار اضافه شده را حذف و برمی‌گرداند (در پایتون +۳.۷)
last_item = user.popitem()
print(f"آخرین آیتم حذف شده با .popitem(): {last_item}")
print(f"دیکشنری بعد از .popitem(): {user}")

# .clear(): تمام اعضای دیکشنری را حذف می‌کند
user.clear()
print(f"بعد از .clear(): {user}")
```

```python
print("--- ۵. پیمایش (Iteration) روی دیکشنری‌ها ---")
product = {
    "name": "Laptop",
    "price": 1500,
    "brand": "Dell",
    "in_stock": True
}
# پیمایش روی کلیدها (روش پیش‌فرض)
print("پیمایش روی کلیدها:")
for key in product:
    print(f"  کلید: {key}, مقدار: {product[key]}")

# پیمایش روی مقادیر با .values()
print("\nپیمایش روی مقادیر:")
for value in product.values():
    print(f"  مقدار: {value}")

# پیمایش روی زوج‌های کلید-مقدار با .items() (بهترین و رایج‌ترین روش)
print("\nپیمایش روی آیتم‌ها (کلید و مقدار):")
for key, value in product.items():
    print(f"  {key}: {value}")
```

```python
print("--- ۶. دیکشنری‌های تو در تو (Nested Dictionaries) ---")
students = {
    'std101': {'name': 'Ali', 'major': 'Computer Science', 'grades': [18, 19, 20]},
    'std102': {'name': 'Sara', 'major': 'Electrical Engineering', 'grades': [19, 17.5, 20]}
}
# دسترسی به داده‌های تو در تو
sara_major = students['std102']['major']
print(f"رشته تحصیلی سارا: {sara_major}")

# تغییر یک مقدار داخلی
students['std101']['grades'].append(17)
print(f"نمرات جدید علی: {students['std101']['grades']}")
```

```python
print("--- ۷. Dictionary Comprehensions (یک روش قدرتمند برای ساخت دیکشنری) ---")
# ساخت یک دیکشنری از مربع اعداد
squares = {x: x*x for x in range(1, 6)}
print(f"دیکشنری مربع اعداد: {squares}")

# ساخت دیکشنری از یک لیست با یک شرط
fruits = ['apple', 'banana', 'cherry']
fruit_lengths = {fruit: len(fruit) for fruit in fruits if len(fruit) > 5}
print(f"دیکشنری میوه‌های با نام طولانی: {fruit_lengths}")
```

```python
print("--- ۸. انواع داده مجاز برای کلیدها ---")
# کلیدهای دیکشنری باید غیرقابل تغییر (immutable) و یکتا (unique) باشند.
# رشته، عدد و تاپل مجاز هستند. لیست و دیکشنری دیگر مجاز نیستند.
misc_dict = {
    "name": "Test",
    123: "This is a number key",
    (10, 20): "This is a tuple key"
}
print(f"دیکشنری با کلیدهای مختلف: {misc_dict}")
print(f"دسترسی با کلید تاپل: {misc_dict[(10, 20)]}")

# تلاش برای استفاده از لیست به عنوان کلید
try:
    invalid_dict = {[1, 2]: "value"}
except TypeError as e:
    print(f"\nخطای مورد انتظار هنگام استفاده از لیست به عنوان کلید: {e}")
```

<div class="card" dir="rtl">
<h2>🔑 مفهوم Immutable در پایتون</h2>

<p>
برخی داده‌ها در پایتون <b>غیرقابل تغییر (Immutable)</b> هستند.  
این یعنی وقتی داده ساخته شد، <b>مقدارش را نمی‌توان مستقیماً تغییر داد</b>.  
هر تغییری در واقع یک نسخه جدید از داده ایجاد می‌کند.
</p>

<h4>مثال با عدد (int)</h4>
<div style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
<pre>
x = 10
y = x
x = x + 5    # یک شی جدید 15 ساخته می‌شود، x به آن اشاره می‌کند
print(x)     # 15
print(y)     # 10 (داده قبلی دست نخورده باقی ماند)

<h4>مثال با رشته (str)</h4>

text = "Hello"

</div>

<div class="card" dir="rtl" style="box-sizing: border-box;">
    <h2>🧠 Mutable vs Immutable - توضیح خودمونی</h2>
    <p>
        بیایید مفهوم mutable و immutable رو با مثال‌های ساده و داستان خودمونی درک کنیم:
    </p>
    <div style="background-color:#fff9c4; color:#333; padding:20px; border-radius:12px; border:2px solid #f0c419; box-shadow: 0 4px 20px rgba(0,0,0,0.2); font-weight:bold; line-height:1.6; box-sizing: border-box;">
        💡 تصور کن هر متغیر مثل یک برچسبه که روی یک شی در حافظه چسبیده.  
        مثال:
        <ul>
            <li>وقتی می‌گی <code>x = 10</code> → برچسب x روی شی عدد ۱۰ چسبیده.</li>
            <li>حالا <code>x = x + 5</code> → عدد قبلی تغییر نمی‌کنه، بلکه یک شی جدید ۱۵ ساخته می‌شه و برچسب x بهش چسبیده. شی ۱۰ هم هنوز همون‌جاست.</li>
            <li>رشته‌ها و تاپل‌ها هم همین‌جور هستن چون <b>غیرقابل تغییر (immutable)</b> هستن.</li>
            <li>اما لیست و دیکشنری <b>قابل تغییر (mutable)</b> هستن، پس وقتی می‌گی <code>lst[0] = 100</code>، خود همون شی موجود تو حافظه تغییر می‌کنه و هیچ شی جدید ساخته نمی‌شه.</li>
            <li>این یعنی هر کسی که به همون لیست اشاره داشته باشه، تغییر رو می‌بینه!</li>
        </ul>
        ⚠️ نکته مهم: وقتی فکر می‌کنید "داده رو تغییر دادم"، اول ببینید آیا واقعا همون شی تغییر کرده یا یه شی جدید ساخته شده و متغیر شما فقط بهش اشاره می‌کنه!
