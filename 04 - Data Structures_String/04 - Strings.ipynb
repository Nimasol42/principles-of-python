    <h2>رشته‌ها (Strings): دنیای متن در پایتون</h2>
    <p>
        به دنیای رشته‌ها خوش آمدید! <b>رشته (String)</b> یکی از بنیادی‌ترین و پراستفاده‌ترین انواع داده در پایتون است. هر متنی که در برنامه خود استفاده می‌کنید، از یک اسم ساده گرفته تا یک پاراگراف طولانی، در قالب یک رشته ذخیره می‌شود.
    </p>
    <p>
        رشته‌ها در واقع دنباله‌ای از کاراکترها (حروف، اعداد، نمادها) هستند. در پایتون، می‌توانیم رشته‌ها را با استفاده از تک کوتیشن (<code>'</code>)، دابل کوتیشن (<code>"</code>) یا سه کوتیشن (<code>'''</code> یا <code>"""</code>) برای متن‌های چندخطی ایجاد کنیم.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
# روش‌های مختلف ساخت رشته
name = 'Ali'
greeting = "Hello, World!"
multiline_text = """This is a string
that spans across
multiple lines."""
print(name)
print(greeting)
print(multiline_text)
    </pre>
    <div class="slide-note">
        <strong>نکته کلیدی:</strong> رشته‌ها در پایتون <b>غیرقابل تغییر (Immutable)</b> هستند. یعنی پس از ساخته شدن، نمی‌توانید یک کاراکتر از آن را مستقیماً تغییر دهید. این ویژگی دقیقاً مانند تاپل‌هاست!
    </div>
</div>

```python
print("--- ۱. روش‌های مختلف ساخت رشته ---")
single_quote = 'سلام پایتون!'
double_quote = "این هم یک رشته است."
# برای متن‌های چند خطی یا شامل کوتیشن
multi_line = """
این یک متن طولانی است
که در چندین خط نوشته شده.
می‌تواند شامل 'تک کوتیشن' یا "دابل کوتیشن" باشد.
"""
print(single_quote)
print(double_quote)
print(multi_line)
print("نوع متغیر single_quote:", type(single_quote))

print("\n--- ۲. اتصال (Concatenation) و تکرار (Repetition) ---")
first_name = "علی"
last_name = "رضایی"
full_name = first_name + " " + last_name
print("نام کامل:", full_name)

separator = "-" * 20  # تکرار رشته برای ساخت جداکننده
print(separator)

print("\n--- ۳. نمایش غیرقابل تغییر بودن (Immutability) ---")
my_string = "Hello"
print("رشته اولیه:", my_string)
try:
    # این خط کد خطا می‌دهد زیرا نمی‌توان یک کاراکتر را مستقیماً تغییر داد
    my_string[0] = 'J'
except TypeError as e:
    print(f"خطا: {e}")
    print("برای تغییر رشته، باید یک رشته جدید بسازیم:")
    new_string = 'J' + my_string[1:]
    print("رشته جدید:", new_string)
```

<div class="card" dir="rtl">
    <h2>دسترسی به کاراکترها: اندیس و برش</h2>
    <p>
        از آنجایی که رشته‌ها یک <b>دنباله</b> هستند، می‌توانیم درست مانند لیست‌ها و تاپل‌ها به اجزای داخلی آن‌ها (کاراکترها) دسترسی پیدا کنیم. این کار از دو طریق اصلی انجام می‌شود:
    </p>
    <h4>۱. اندیس‌گذاری (Indexing)</h4>
    <p>
        هر کاراکتر در رشته یک شماره یا <b>اندیس</b> دارد. اندیس‌ها از صفر شروع می‌شوند. همچنین می‌توان از اندیس‌های منفی برای شمارش از انتهای رشته استفاده کرد.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
text = "Python"
# P  y  t  h  o  n
# 0  1  2  3  4  5
#-6 -5 -4 -3 -2 -1
print(text[0])   # خروجی: 'P' (اولین کاراکتر)
print(text[3])   # خروجی: 'h'
print(text[-1])  # خروجی: 'n' (آخرین کاراکتر)
    </pre>
    <h4>۲. برش زدن (Slicing)</h4>
    <p>
        با استفاده از برش، می‌توانیم یک زیررشته (substring) یا بخشی از رشته اصلی را استخراج کنیم. ساختار کلی آن به صورت <code>[start:stop:step]</code> است.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
message = "Hello, World!"
print(message[0:5])     # خروجی: 'Hello' (از اندیس 0 تا قبل از 5)
print(message[7:])      # خروجی: 'World!' (از اندیس 7 تا انتها)
print(message[:5])      # خروجی: 'Hello' (از ابتدا تا قبل از 5)
print(message[::2])     # خروجی: 'Hlo ol!' (یک در میان)
print(message[::-1])    # یک تکنیک جالب برای برعکس کردن رشته
# خروجی: '!dlroW ,olleH'
    </pre>
</div>

```python
text = "PYTHON IS AWESOME"
print(f"رشته مورد بررسی: '{text}'")
print("-" * 20)

print("--- ۱. اندیس‌گذاری (Indexing) ---")
print(f"کاراکتر اول (اندیس 0): {text[0]}")
print(f"کاراکتر هفتم (اندیس 6): {text[6]}") # کاراکتر فاصله
print(f"آخرین کاراکتر (اندیس -1): {text[-1]}")
print(f"سومین کاراکتر از آخر (اندیس -3): {text[-3]}")

# تلاش برای دسترسی به یک اندیس خارج از محدوده
try:
    print(text[100])
except IndexError as e:
    print(f"\nخطای مورد انتظار: {e}")


print("\n--- ۲. برش زدن (Slicing) ---")
# فرمت کلی: [start:stop:step]

print(f"از اندیس 0 تا قبل از 6: '{text[0:6]}'")    # PYTHON
print(f"از اندیس 7 تا انتها: '{text[7:]}'")       # IS AWESOME
print(f"از ابتدا تا قبل از 6: '{text[:6]}'")       # PYTHON
print(f"کل رشته: '{text[:]}'")

print("\n--- ۳. برش با گام (Step) ---")
print(f"یک در میان کاراکترها: '{text[::2]}'") # PTO SWSM
print(f"برعکس کردن رشته: '{text[::-1]}'")

# یک مثال پیچیده‌تر
# برعکس کردن کلمه AWESOME
# اندیس AWESOME از 11 تا 17 است
print(f"برعکس کردن کلمه AWESOME: '{text[17:10:-1]}'")
```

    <h2>متدهای پرکاربرد: تغییر و پاک‌سازی</h2>
    <p>
        رشته‌ها در پایتون دارای متدهای داخلی بسیار قدرتمندی هستند که به ما اجازه می‌دهند عملیات‌های مختلفی را روی آن‌ها انجام دهیم. به یاد داشته باشید که چون رشته‌ها immutable هستند، این متدها رشته اصلی را تغییر <b>نمی‌دهند</b>، بلکه یک <b>رشته جدید</b> را به عنوان خروجی برمی‌گردانند.
    </p>
    <ul>
        <li><code>s.upper()</code> و <code>s.lower()</code>: برای تبدیل تمام حروف به بزرگ یا کوچک.</li>
        <li><code>s.capitalize()</code> و <code>s.title()</code>: برای بزرگ کردن حرف اول جمله یا حرف اول تمام کلمات.</li>
        <li><code>s.strip()</code>: برای حذف فاصله‌های خالی (whitespace) از ابتدا و انتهای رشته. نسخه‌های <code>lstrip()</code> و <code>rstrip()</code> نیز برای حذف از چپ یا راست وجود دارند.</li>
        <li><code>s.replace(old, new)</code>: برای جایگزین کردن یک زیررشته با زیررشته‌ای دیگر.</li>
    </ul>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
text = "   hello python   "
title_text = "welcome to the jungle"
# تغییر حروف
print(text.upper()) # خروجی: '   HELLO PYTHON   '
print(title_text.title()) # خروجی: 'Welcome To The Jungle'
# پاک‌سازی فاصله‌ها
cleaned_text = text.strip()
print(f"'{cleaned_text}'") # خروجی: "'hello python'"
# جایگزینی
replaced_text = cleaned_text.replace("python", "world")
print(replaced_text) # خروجی: 'hello world'
    </pre>
</div>

```python
print("--- ۱. تبدیل حروف (Case Conversion) ---")
mixed_case = "pyThoN PrOGRAmmINg"
print(f"رشته اولیه: '{mixed_case}'")
print(f"با حروف بزرگ (upper): '{mixed_case.upper()}'")
print(f"با حروف کوچک (lower): '{mixed_case.lower()}'")
print(f"حرف اول جمله بزرگ (capitalize): '{mixed_case.capitalize()}'")
print(f"حرف اول تمام کلمات بزرگ (title): '{mixed_case.title()}'")


print("\n--- ۲. پاک‌سازی فاصله‌ها (Stripping) ---")
messy_text = "    Some extra spaces    "
print(f"رشته آلوده: '{messy_text}'")
print(f"حذف از دو طرف (strip): '{messy_text.strip()}'")
print(f"حذف فقط از چپ (lstrip): '{messy_text.lstrip()}'")
print(f"حذف فقط از راست (rstrip): '{messy_text.rstrip()}'")

# حذف کاراکترهای دیگر
messy_chars = "!!!Python!!!"
print(f"\nرشته با کاراکترهای اضافه: '{messy_chars}'")
print(f"حذف علامت تعجب: '{messy_chars.strip('!')}'")


print("\n--- ۳. جایگزینی (Replacing) ---")
sentence = "I like cats. I think cats are cute."
print(f"جمله اولیه: '{sentence}'")
# جایگزینی تمام موارد
new_sentence = sentence.replace("cats", "dogs")
print(f"جایگزینی 'cats' با 'dogs': '{new_sentence}'")

# جایگزینی فقط اولین مورد
limited_replace = sentence.replace("cats", "dogs", 1)
print(f"جایگزینی فقط اولین 'cats': '{limited_replace}'")
```

    <h2>متدهای پرکاربرد: تقسیم و اتصال</h2>
    <p>
        دو متد بسیار مهم <code>split()</code> و <code>join()</code> پل ارتباطی بین رشته‌ها و لیست‌ها هستند و در پردازش متن کاربرد فراوانی دارند.
    </p>
    <h4><code>s.split(separator)</code></h4>
    <p>
        این متد یک رشته را بر اساس یک جداکننده (separator) به لیستی از رشته‌ها تقسیم می‌کند. اگر جداکننده مشخص نشود، بر اساس فاصله خالی تقسیم می‌کند.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
sentence = "this is a sample sentence"
csv_data = "Ali,Rezaei,30"
words = sentence.split()
print(words) # خروجی: ['this', 'is', 'a', 'sample', 'sentence']
user_info = csv_data.split(',')
print(user_info) # خروجی: ['Ali', 'Rezaei', '30']
    </pre>
    <h4><code>separator.join(iterable)</code></h4>
    <p>
        این متد برعکس <code>split()</code> عمل می‌کند. اعضای یک iterable (مانند لیست) را با استفاده از یک رشته جداکننده به هم متصل کرده و یک رشته واحد تولید می‌کند.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
word_list = ['Python', 'is', 'awesome']
date_parts = ['2023', '10', '26']
# اتصال با فاصله
full_sentence = " ".join(word_list)
print(full_sentence) # خروجی: 'Python is awesome'
# اتصال با خط تیره
formatted_date = "-".join(date_parts)
print(formatted_date) # خروجی: '2023-10-26'
    </pre>
</div>

```python
print("--- ۱. تقسیم رشته (split) ---")
# تقسیم بر اساس فاصله
phrase = "Python is a versatile language"
words = phrase.split()
print(f"جمله: '{phrase}'")
print(f"لیست کلمات: {words}")

# تقسیم بر اساس یک جداکننده مشخص
csv_line = "John,Doe,42,New York"
user_data = csv_line.split(',')
print(f"\nخط CSV: '{csv_line}'")
print(f"داده‌های کاربر: {user_data}")

# محدود کردن تعداد تقسیم‌ها
url = "https://example.com/path/to/resource"
parts = url.split('/', 3) # حداکثر ۳ تقسیم
print(f"\nURL: '{url}'")
print(f"تقسیم با محدودیت: {parts}")


print("\n--- ۲. اتصال لیست به رشته (join) ---")
word_list = ["Learning", "Python", "is", "fun"]
separator = " "
sentence = separator.join(word_list)
print(f"لیست اولیه: {word_list}")
print(f"جمله ساخته شده: '{sentence}'")

# با یک جداکننده دیگر
filename_parts = ["document", "2023", "final"]
filename = "_".join(filename_parts) + ".pdf"
print(f"\nاجزای نام فایل: {filename_parts}")
print(f"نام فایل نهایی: '{filename}'")

# خطا هنگام تلاش برای اتصال لیستی با اعضای غیررشته
mixed_list = ["Item", 123]
try:
    " ".join(mixed_list)
except TypeError as e:
    print(f"\nخطای مورد انتظار: {e}")
    # راه حل: تبدیل همه اعضا به رشته
    fixed_list = [str(item) for item in mixed_list]
    print(f"راه حل: {' '.join(fixed_list)}")
```

    <h2>قالب‌بندی رشته‌ها (Formatting)</h2>
    <p>
        اغلب نیاز داریم متغیرها را داخل یک رشته قرار دهیم. برای این کار روش‌های مختلفی وجود دارد، اما <b>f-string</b> مدرن‌ترین و خواناترین روش است.
    </p>
    <h4>۱. روش قدیمی (اپراتور <code>%</code>)</h4>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
name = "Sara"
age = 28
print("Name: %s, Age: %d" % (name, age))
    </pre>
    <h4>۲. متد <code>.format()</code></h4>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
print("Name: {}, Age: {}".format(name, age))
# یا با استفاده از اندیس
print("Name: {0}, Age: {1}".format(name, age))
    </pre>
    <h4>۳. f-Strings (رویکرد مدرن و پیشنهادی)</h4>
    <p>
        این روش که از پایتون 3.6 به بعد معرفی شد، بسیار خوانا و کارآمد است. کافیست قبل از کوتیشن حرف <code>f</code> را قرار دهید و متغیرها را مستقیماً داخل آکولاد <code>{}</code> بنویسید.
    </p>
    <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
product = "Laptop"
price = 1500
tax = 0.09
print(f"The {product} costs ${price}.")
# حتی می‌توانید عبارات پایتون را مستقیماً داخل آن اجرا کنید!
print(f"Total price with tax: ${price * (1 + tax):.2f}") # .2f برای نمایش دو رقم اعشار
    </pre>
</div>

```python
name = "Alice"
age = 30
balance = 12345.6789

print("--- ۱. روش قدیمی (اپراتور %) ---")
print("Name: %s, Age: %d, Balance: %.2f" % (name, age, balance))

print("\n--- ۲. متد .format() ---")
print("Name: {}, Age: {}, Balance: {:.2f}".format(name, age, balance))
print("Name: {n}, Age: {a}".format(n=name, a=age)) # با کلیدواژه

print("\n--- ۳. f-Strings (رویکرد مدرن) ---")
# تزریق ساده متغیر
print(f"User '{name}' is {age} years old.")

# اجرای عبارات پایتون در داخل f-string
print(f"In 5 years, {name} will be {age + 5} years old.")
print(f"Name in uppercase: {name.upper()}")

# قالب‌بندی اعداد
print(f"Balance rounded to 2 decimal places: ${balance:.2f}")
print(f"Balance with thousand separators: ${balance:,.2f}")

# تراز کردن متن
# < (چپ‌چین), ^ (وسط‌چین), > (راست‌چین)
print(f"'{name:<10}'") # تراز چپ در 10 کاراکتر
print(f"'{name:^10}'") # تراز وسط
print(f"'{name:>10}'") # تراز راست

# یک قابلیت عالی برای دیباگ کردن (از پایتون ۳.۸ به بعد)
print(f"مقادیر متغیرها: {name=}, {age=}")
```

    <h2>جمع‌بندی: رشته‌ها در اکوسیستم پایتون</h2>
    <p>
        رشته‌ها به تنهایی قدرتمند هستند، اما قدرت واقعی آن‌ها زمانی مشخص می‌شود که با دیگر ساختارهای داده در پایتون تعامل می‌کنند.
    </p>
    <ul>
        <li><b>رشته و لیست:</b> می‌توان به راحتی با <code>split()</code> رشته را به لیست و با <code>join()</code> لیست را به رشته تبدیل کرد. تابع <code>list()</code> یک رشته را به لیستی از کاراکترها تبدیل می‌کند.</li>
        <li><b>رشته و تاپل:</b> رشته‌ها مانند تاپل‌ها غیرقابل تغییر (immutable) هستند. تابع <code>tuple()</code> یک رشته را به تاپلی از کاراکترها تبدیل می‌کند.</li>
        <li><b>رشته و دیکشنری:</b> رشته‌ها معمولاً به عنوان کلید (key) در دیکشنری‌ها استفاده می‌شوند، زیرا immutable هستند.</li>
        <li><b>رشته و حلقه‌ها:</b> می‌توان به سادگی روی کاراکترهای یک رشته پیمایش کرد:
            <pre style="background-color:#272822; color:#f8f8f2; padding:10px; border-radius:8px; direction:ltr; text-align:left;">
for char in "salam":
    print(char, end='-')
# خروجی: s-a-l-a-m-
            </pre>
        </li>
    </ul>
    <div class="slide-note">
        <strong>یادآوری نهایی:</strong> همیشه به یاد داشته باشید که متدهای رشته، رشته اصلی را تغییر نمی‌دهند. اگر می‌خواهید نتیجه را ذخیره کنید، باید آن را در یک متغیر جدید بریزید. <br>
        <code>my_string = my_string.upper()</code>

```python
text = "the quick brown fox jumps over the lazy dog"
print(f"رشته مورد بررسی: '{text}'")
print("-" * 20)

print("--- ۱. بررسی وجود زیررشته (in) ---")
print(f"آیا 'fox' در رشته وجود دارد؟ {'fox' in text}")
print(f"آیا 'cat' در رشته وجود دارد؟ {'cat' in text}")

print("\n--- ۲. پیدا کردن موقعیت زیررشته (find و index) ---")
# .find() در صورت پیدا نکردن -1 برمی‌گرداند
print(f"موقعیت اولین 'the': {text.find('the')}")
print(f"موقعیت 'fox': {text.find('fox')}")
print(f"موقعیت 'cat': {text.find('cat')}") # پیدا نشد

# .index() در صورت پیدا نکردن خطا می‌دهد
print(f"\nموقعیت 'lazy' با index: {text.index('lazy')}")
try:
    text.index('cat')
except ValueError as e:
    print(f"خطای مورد انتظار از index: {e}")

print("\n--- ۳. شمارش تعداد تکرار (count) ---")
print(f"تعداد تکرار 'the': {text.count('the')}")
print(f"تعداد تکرار حرف 'o': {text.count('o')}")

print("\n--- ۴. بررسی ابتدا و انتهای رشته (startswith و endswith) ---")
filename = "report.pdf"
print(f"نام فایل: '{filename}'")
print(f"آیا با 'report' شروع می‌شود؟ {filename.startswith('report')}")
print(f"آیا به '.docx' ختم می‌شود؟ {filename.endswith('.docx')}")

print("\n--- ۵. بررسی نوع کاراکترهای رشته ---")
print(f"'12345'.isdigit(): {'12345'.isdigit()}") # آیا تماما عدد است؟
print(f"'Python'.isalpha(): {'Python'.isalpha()}") # آیا تماما حروف الفبا است؟
print(f"'Python3'.isalnum(): {'Python3'.isalnum()}") # آیا تماما حروف و عدد است؟
print(f"'   '.isspace(): {'   '.isspace()}") # آیا تماما فاصله خالی است؟
print(f"'Title Case'.istitle(): {'Title Case'.istitle()}") # آیا فرمت تایتل دارد؟
```
