<div class="card" dir="rtl">
<h2>متغیر (Variable)</h2>
<!-- توضیحات اصلی -->
<p>
در پایتون، متغیر یک نام است که به یک مقدار در حافظه کامپیوتر اشاره می‌کند. به زبان ساده‌تر، متغیر همان «جعبه برچسب‌دار» ماست.
</p>
<p>
شما یک نام (برچسب) انتخاب می‌کنید و یک مقدار (محتویات) را داخل آن قرار می‌دهید. از آن به بعد، هر جا از آن نام استفاده کنید، پایتون به مقداری که داخل آن ذخیره شده است، مراجعه می‌کند.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
برای ساختن یک متغیر، کافی است یک نام برای آن انتخاب کرده و با علامت <span style="color:blue;">=</span> یک مقدار به آن اختصاص دهید.
</p>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/assignment.png" alt="Variable Assignment" style="max-width: 400px;">
</div>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
# score نام متغیر است
# 100 مقداری است که در آن ذخیره می‌شود
score = 100
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در کد بالا، ما یک متغیر به نام <code>score</code> ساخته‌ایم و مقدار عددی <code>100</code> را در آن ذخیره کرده‌ایم. به عمل تخصیص مقدار به متغیر، <strong>انتساب (Assignment)</strong> می‌گویند.
</p>
</div>

<div class="card" dir="rtl">
<h2>چطور از متغیرها استفاده کنیم؟</h2>
<!-- توضیحات اصلی -->
<p>
بعد از اینکه یک مقدار را در متغیر ذخیره کردید، می‌توانید به راحتی از آن استفاده کنید. کافی است نام آن را صدا بزنید!
</p>
<p>
برای مثال، می‌توانیم مقداری که در متغیر <code>score</code> ذخیره کرده‌ایم را با دستور <code>print()</code> روی صفحه نمایش دهیم.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
وقتی از نام متغیر استفاده می‌کنید، پایتون به صورت خودکار <span style="color:green;">مقدار داخل آن</span> را جایگزین می‌کند.
</p>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/using_variable.png" alt="Using Variables" style="max-width: 380px;">
</div>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
score = 100
print(score)
player_name = "Ali"
print(player_name)
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
اگر کد بالا را اجرا کنید، در خروجی ابتدا عدد <code>100</code> و سپس نام <code>Ali</code> نمایش داده می‌شود. ما می‌توانیم انواع مختلفی از داده‌ها مثل اعداد، متن‌ها و... را در متغیرها ذخیره کنیم.
</p>
</div>

```python
score = 100
print(score)
player_name = "Ali"
print(player_name)
```

<div class="card" dir="rtl">
<h2>تغییر مقدار یک متغیر</h2>
<!-- توضیحات اصلی -->
<p>
یکی از ویژگی‌های مهم متغیرها این است که مقدار آن‌ها می‌تواند در طول اجرای برنامه تغییر کند. به همین دلیل به آن‌ها «متغیر» می‌گویند!
</p>
<p>
فرض کنید بازیکن در بازی ما ۱۰ امتیاز دیگر کسب می‌کند. ما به راحتی می‌توانیم مقدار جدید را در همان متغیر قبلی ذخیره کنیم.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
هر بار که مقدار جدیدی به یک متغیر اختصاص می‌دهید، مقدار <span style="color:red;">قبلی</span> از بین می‌رود و مقدار <span style="color:green;">جدید</span> جایگزین آن می‌شود.
</p>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/reassign_var.png" alt="Changing Variable Value" style="max-width: 420px;">
</div>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
score = 100
print("امتیاز اولیه:", score)
# بازیکن ۱۰ امتیاز جدید می‌گیرد
score = 110
print("امتیاز جدید:", score)
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در این مثال، ابتدا مقدار <code>100</code> در متغیر <code>score</code> قرار دارد، اما در خط بعدی، ما آن را به <code>110</code> تغییر می‌دهیم. کامپیوتر همیشه آخرین مقداری که به یک متغیر اختصاص داده شده را به خاطر می‌سپارد.
</p>
</div>

```python
score = 100
print("امتیاز اولیه:", score)
# بازیکن ۱۰ امتیاز جدید می‌گیرد
score = 110
print("امتیاز جدید:", score)
```

درست مثل اینکه ما برای نامگذاری فرزندانمان از هر اسم عجیب و غریبی استفاده نمی‌کنیم، در برنامه‌نویسی هم برای نامگذاری متغیرها باید از یک سری قوانین پیروی کنیم.
</p>
<p>
رعایت این قوانین باعث می‌شود هم پایتون کد ما را بفهمد و هم برنامه‌نویسان دیگر (و حتی خود ما در آینده!) بتوانند به راحتی کد را بخوانند.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
نام یک متغیر <span style="color:green;">باید</span> با یک حرف یا آندرلاین (<code>_</code>) شروع شود و <span style="color:red;">نمی‌تواند</span> با عدد آغاز گردد.
</p>

_player_id = 1234

<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">

<h2>انتساب (Assignment) چیست؟</h2>
<!-- توضیحات اصلی -->
<p>
همانطور که دیدیم، «انتساب» به عمل قرار دادن یک مقدار در یک متغیر گفته می‌شود. این کار همیشه با استفاده از عملگر مساوی (<code>=</code>) انجام می‌شود.
</p>
<p>
مهم است بدانید که این عملگر با «مساوی بودن» در ریاضیات تفاوت دارد. در پایتون، علامت <code>=</code> یعنی: «مقداری که در سمت راست قرار دارد را بردار و داخل متغیری که در سمت چپ است، بریز.»
</p>
<p>
سمت راست می‌تواند یک عدد، یک متن، نتیجه یک محاسبه یا حتی یک متغیر دیگر باشد.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
عملگر <code>=</code> همیشه مقدار <span style="color:blue;">راست</span> را در متغیر <span style="color:red;">چپ</span> قرار می‌دهد.
</p>

<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
# انتساب یک عدد
x = 10 </br>
# انتساب یک متن
message = "Hello, Python!" </br>
# انتساب نتیجه یک محاسبه
total_price = 5 * 1200 </br>
# انتساب مقدار یک متغیر به متغیر دیگر
y = x
print(y) # --> خروجی: 10
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در مثال آخر، ابتدا مقدار متغیر <code>x</code> (که ۱۰ است) خوانده شده و سپس همان مقدار در متغیر جدید <code>y</code> کپی می‌شود. بنابراین، هر دو متغیر مقدار ۱۰ را خواهند داشت.

<h2>انتساب چندگانه (Multiple Assignment)</h2>
<!-- توضیحات اصلی -->
<p>
پایتون به ما اجازه می‌دهد تا کارها را به صورت خلاصه‌تر و تمیزتر انجام دهیم. یکی از این قابلیت‌ها، انتساب یک مقدار به چندین متغیر به صورت همزمان است.
</p>
<p>
اگر بخواهید چندین متغیر در ابتدای برنامه مقدار یکسانی داشته باشند (مثلاً صفر)، نیازی نیست آن‌ها را در خطوط جداگانه بنویسید.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
می‌توانید چندین متغیر را با عملگر <code>=</code> به هم زنجیر کرده و در انتها یک مقدار به همه‌ی آن‌ها اختصاص دهید.
</p>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">

print(x) # --> خروجی: 0

<!-- مثال توضیحی -->

<p>

<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">

<!-- کد نمونه -->

<h2>عملگرهای انتساب ترکیبی</h2>
<!-- توضیحات اصلی -->
<p>
در برنامه‌نویسی بسیار پیش می‌آید که بخواهیم مقدار یک متغیر را بر اساس مقدار فعلی خودش تغییر دهیم. برای مثال، ۱۰ واحد به امتیاز یک بازیکن اضافه کنیم.
</p>
<p>
به جای اینکه بنویسیم <code>score = score + 10</code>، پایتون راه کوتاه‌تری پیش پای ما گذاشته است که به آن <strong>عملگرهای انتساب ترکیبی</strong> (Augmented Assignment) می‌گویند.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
این عملگرها یک عملیات ریاضی را با انتساب ترکیب می‌کنند و کد را تمیزتر و کوتاه‌تر می‌سازند.
</p>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
score = 100 </br>
# اضافه کردن ۱۰ به امتیاز
score += 10 # معادل: score = score + 10
print(score) # --> خروجی: 110 </br>
# کم کردن ۵ از امتیاز
score -= 5 # معادل: score = score - 5
print(score) # --> خروجی: 105 </br>
# دو برابر کردن امتیاز
score *= 2 # معادل: score = score * 2
print(score) # --> خروجی: 210
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
<strong>نکات کلیدی:</strong><br>
<ul>
<li style="margin-right: 20px;">این عملگرها برای تمام عملیات اصلی ریاضی وجود دارند: <code>+=</code>, <code>-=</code>, <code>*=</code>, <code>/=</code>, <code>%=</code> و <code>**=</code>.</li>
<li style="margin-right: 20px;"><strong>نکته مهم:</strong> پایتون همیشه ابتدا عبارت سمت راست (مثلاً <code>score + 10</code>) را با استفاده از <strong>آخرین مقدار موجود</strong> در متغیر <code>score</code> محاسبه کرده و سپس نتیجه نهایی را در خود متغیر ذخیره می‌کند.</li>
<li style="margin-right: 20px;">استفاده از این عملگرها یک عادت خوب در برنامه‌نویسی پایتون محسوب می‌شود، چون کد را خواناتر و کوتاه‌تر می‌کند.</li>
</ul>
</p>
</div>

```python
# شروع با یک مقدار اولیه
x = 10
print("مقدار اولیه:", x)

# جمع با انتساب
x += 5
print("بعد از x += 5 :", x)

# تفریق با انتساب
x -= 3
print("بعد از x -= 3 :", x)

# ضرب با انتساب
x *= 2
print("بعد از x *= 2 :", x)

# تقسیم با انتساب
x /= 4
print("بعد از x /= 4 :", x)

# توان با انتساب
x **= 2
print("بعد از x **= 2 :", x)

# باقیمانده با انتساب
x %= 5
print("بعد از x %= 5 :", x)
```

<h3 style="text-align: right; color: #007bff;">۱. رشته‌ای که شبیه عدد است، عدد نیست!</h3>
<p>
مهم‌ترین قانون این است: <strong>هر چیزی که داخل علامت نقل قول باشد، یک رشته است.</strong> حتی اگر محتوای آن فقط از اعداد تشکیل شده باشد. پایتون با چنین داده‌ای مانند متن رفتار می‌کند، نه یک مقدار عددی.
</p>
<p>
<b>آنالوژی:</b> فکر کنید روی یک جعبه، برچسب "123" را چسبانده‌اید. این برچسب فقط یک نوشته است و به معنای وجود ۱۲۳ شیء در جعبه نیست.
</p>

numeric_value = 123
string_value = "123" </br>
print("نوع داده اول:", type(numeric_value)) # --> خروجی: <class 'int'>
print("نوع داده دوم:", type(string_value)) # --> خروجی: <class 'str'> </br>

<h3 style="text-align: right; color: #dc3545;">۲. تفاوت رشته خالی و رشته حاوی فاصله (Space)</h3>
<p>
یک رشته خالی (<code>""</code>) هیچ کاراکتری ندارد (طول آن صفر است). اما یک رشته که فقط یک یا چند فاصله (space) دارد، خالی نیست! <strong>فاصله نیز یک کاراکتر محسوب می‌شود.</strong>
</p>
<p>
<b>آنالوژی:</b> یک پاکت نامه خالی در مقابل یک پاکت نامه که فقط یک کاغذ سفید تا شده در آن است. هر دو از بیرون شبیه هم به نظر می‌رسند، اما یکی واقعاً خالی است و دیگری محتوا دارد.
</p>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
empty_string = ""
space_string = " " </br>
# تابع len() تعداد کاراکترهای یک رشته را برمی‌گرداند
print("طول رشته خالی:", len(empty_string)) # --> خروجی: 0
print("طول رشته فاصله:", len(space_string)) # --> خروجی: 1
</pre>

<h3 style="text-align: right; color: #28a745;">۳. نقل قول تکی (' ') یا جفتی (" ")؟</h3>

</p>
<p>
انتخاب بین این دو معمولاً به محتوای خود رشته بستگی دارد. اگر در متن شما از آپاستروف (که همان نقل قول تکی است) استفاده شده، بهتر است کل رشته را با نقل قول جفتی بسازید و برعکس.
</p>

<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
# اینجا چون از آپاستروف استفاده شده، از نقل قول جفتی استفاده می‌کنیم
message1 = "It's a beautiful day." </br>
# اینجا چون از نقل قول جفتی در متن استفاده شده، از نقل قول تکی استفاده می‌کنیم

<h2>نوع داده: بولین (Boolean)</h2>
<!-- توضیحات اصلی -->
<p>
نوع داده <strong>بولین</strong> یا <code>bool</code> یکی از ساده‌ترین و در عین حال بنیادی‌ترین انواع داده است. این نوع داده فقط و فقط می‌تواند دو مقدار داشته باشد: <code>True</code> (درست) یا <code>False</code> (نادرست).
</p>
<p>
از مقادیر بولین برای نمایش وضعیت‌های دوتایی (مانند روشن/خاموش، بله/خیر، فعال/غیرفعال) و همچنین برای ذخیره نتیجه عبارات مقایسه‌ای استفاده می‌شود.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
یک بولین، مانند یک کلید برق، فقط دو حالت دارد: <span style="color:green;">True</span> یا <span style="color:red;">False</span>.
</p>

<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
# متغیری برای ذخیره وضعیت ورود کاربر
is_logged_in = True </br>
# متغیری برای ذخیره وضعیت پایان بازی
is_game_over = False </br>
# توجه کنید که حرف اول True و False باید بزرگ باشد
# is_active = true # <- این کد خطا می‌دهد </br>
print(type(is_logged_in))
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
خروجی کد بالا <code>&lt;class 'bool'&gt;</code> خواهد بود. در اسلایدهای بعدی (مبحث عملگرهای مقایسه‌ای) خواهید دید که چطور عباراتی مانند <code>5 > 2</code> به صورت خودکار یک مقدار بولین (در این مورد <code>True</code>) تولید می‌کنند که اساس تصمیم‌گیری در برنامه‌نویسی است.

<h2>مقدمه‌ای بر تبدیل نوع داده (Type Casting)</h2>
<!-- توضیحات اصلی -->
<p>
تا اینجا با انواع مختلف داده‌ها آشنا شدیم. اما آیا می‌توانیم این انواع را با هم ترکیب کنیم؟ مثلاً یک عدد را با یک متن جمع کنیم؟ پاسخ منفی است! پایتون قوانین سخت‌گیرانه‌ای برای این کار دارد.
</p>
<p>
اینجاست که مفهوم <strong>تبدیل نوع داده</strong> یا <strong>Type Casting</strong> (که به آن Type Conversion هم می‌گویند) اهمیت پیدا می‌کند. این فرآیند به معنای تبدیل یک مقدار از یک نوع داده به نوع داده دیگر است.
</p>
<p>
<b>آنالوژی:</b> این فرآیند شبیه تبدیل حالت‌های مختلف آب است. یخ، آب مایع و بخار همگی از یک ماده (H₂O) ساخته شده‌اند، اما در حالت‌های متفاوتی قرار دارند و کاربردهایشان فرق می‌کند. ما با تغییر دما، حالت آن‌ها را برای کاربرد مورد نظرمان تغییر می‌دهیم. در برنامه‌نویسی نیز ما نوع داده‌ها را برای انجام عملیات مورد نظرمان تبدیل می‌کنیم.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
ما داده‌ها را تبدیل می‌کنیم تا بتوانیم عملیات مورد نظر خود را به درستی روی آن‌ها انجام دهیم.
</p>

<p style="text-align: right; line-height: 1.7;">
<b>چرا به تبدیل نوع نیاز داریم؟ دو دلیل اصلی:</b><br>
۱. <strong>برای انجام محاسبات:</strong> ورودی‌ای که از کاربر با تابع <code>input()</code> می‌گیریم، <strong>همیشه</strong> یک رشته است. برای انجام محاسبات ریاضی روی آن، باید ابتدا آن را به یک نوع عددی (مثل <code>int</code>) تبدیل کنیم.<br>
۲. <strong>برای نمایش خروجی:</strong> برای چسباندن یک عدد به یک رشته جهت نمایش در خروجی، ابتدا باید آن عدد را به رشته تبدیل کنیم.
</p>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در پایتون، این تبدیل به دو شکل کلی انجام می‌شود که در اسلایدهای بعدی خواهیم دید:<br>
<ul>
<li style="margin-right: 20px;"><strong>تبدیل ضمنی (Implicit):</strong> پایتون به صورت خودکار این کار را انجام می‌دهد.</li>
<li style="margin-right: 20px;"><strong>تبدیل صریح (Explicit):</strong> ما به صورت دستی و با دستور مستقیم این کار را انجام می‌دهیم.</li>
</ul>
</p>
</div>

<h2>تبدیل نوع ضمنی: جادوی خودکار پایتون</h2>
<!-- توضیحات اصلی -->
<p>
گاهی اوقات، پایتون به قدری هوشمند است که خودش به صورت خودکار نوع یک داده را به نوع دیگری تبدیل می‌کند تا از بروز خطا یا از دست رفتن اطلاعات جلوگیری کند. به این فرآیند <strong>تبدیل نوع ضمنی</strong> یا اتوماتیک می‌گویند.
</p>
<p>
این اتفاق معمولاً در حین انجام عملیات حسابی بین دو نوع داده عددی مختلف (مانند یک <code>int</code> و یک <code>float</code>) رخ می‌دهد.
</p>
<p>
<b>آنالوژی:</b> تصور کنید می‌خواهید ۵ دلار (یک عدد کامل) را با ۲.۵ دلار (یک عدد اعشاری) جمع کنید. ذهن شما به طور خودکار نتیجه را به صورت اعشاری (۷.۵ دلار) محاسبه می‌کند تا بخش اعشاری از بین نرود. پایتون هم دقیقاً همین کار را انجام می‌دهد.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
پایتون برای جلوگیری از، از دست رفتن داده، مقادیر را به نوع داده <span style="color:green;">«دقیق‌تر»</span> یا <span style="color:green;">«کلی‌تر»</span> تبدیل می‌کند.
</p>

<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
integer_number = 10
float_number = 5.5 </br>
# جمع یک عدد صحیح و یک عدد اعشاری
result = integer_number + float_number </br>
print("مقدار نتیجه:", result)
print("نوع داده نتیجه:", type(result))
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در کد بالا، خروجی <code>15.5</code> و نوع داده آن <code>&lt;class 'float'&gt;</code> خواهد بود. پایتون قبل از انجام عمل جمع، متغیر <code>integer_number</code> را به صورت موقت به یک عدد اعشاری (<code>10.0</code>) تبدیل کرد تا بتواند آن را با <code>float_number</code> جمع کند و هیچ بخشی از اطلاعات (یعنی <code>.5</code>) از بین نرود.
</p>

<h2>عبارت (Expression) چیست؟</h2>
<!-- توضیحات اصلی -->
<p>
در زبان پایتون، «عبارت» یا «Expression» ترکیبی از مقادیر، متغیرها و عملگرهاست.
</p>
<p>
می‌توانید به یک عبارت به عنوان یک «مسئله» نگاه کنید که پایتون آن را می‌خواند و حل می‌کند. کامپیوتر هر عبارت را محاسبه (Evaluate) کرده و آن را به یک مقدار واحد و نهایی تبدیل می‌کند.
</p>
<p>
یک عبارت می‌تواند به سادگی یک عدد تنها یا به پیچیدگی یک فرمول ریاضی باشد.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
هر چیزی در کد که پایتون بتواند آن را محاسبه کند و به <span style="color:green;">یک نتیجه واحد</span> برساند، یک عبارت است.
</p>
<div style="text-align: center; margin-bottom: 20px;">
<img src="img/expression.jpg" alt="Expressions" style="max-width: 450px;">
</div>
<!-- کد نمونه -->
<pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
# 5 یک عبارت است که به مقدار 5 ارزیابی می‌شود.
# ---------------------------------------------
# 2 + 3 یک عبارت است که به مقدار 5 ارزیابی می‌شود.
result = 2 + 3
# ---------------------------------------------
# score > 50 یک عبارت است که به True یا False ارزیابی می‌شود.
score = 100
is_high_score = (score > 50)
</pre>
<!-- مثال توضیحی -->
<p style="background-color: #f9f9f9; padding: 10px; border-radius: 8px;">
در مثال‌های بالا، <code>2 + 3</code> یک مسئله است که پایتون آن را حل کرده و به نتیجه <code>5</code> می‌رسد. سپس این نتیجه نهایی در متغیر <code>result</code> ذخیره می‌شود. حتی یک متغیر یا یک عدد تنها نیز یک عبارت محسوب می‌شود.
</p>

<h2>عملگرها (Operators): ابزارهای حل مسئله</h2>
<!-- توضیحات اصلی -->
<p>
اگر عبارت‌ها «مسئله» باشند، <strong>عملگرها</strong> یا <strong>Operators</strong> ابزارها و فعل‌های ما برای حل آن مسئله هستند. عملگرها نمادهای خاصی هستند که به پایتون می‌گویند چه عملیاتی را روی مقادیر و متغیرها (که به آن‌ها عملوند یا Operand می‌گوییم) انجام دهد.
</p>
<p>
برای مثال، در عبارت <code>5 + 3</code>، اعداد ۵ و ۳ عملوند و علامت <code>+</code> عملگر است.
</p>
<p>
عملگرها در پایتون به دسته‌های مختلفی تقسیم می‌شوند که در ادامه هر کدام را بررسی خواهیم کرد.
</p>
<p style="font-size: 1.2em; font-weight: bold; text-align: center; margin: 15px 0;">
عملگرها به پایتون می‌گویند <span style="color:blue;">چه کاری</span> با داده‌ها انجام دهد: آن‌ها را جمع کند، مقایسه کند یا با هم ترکیب کند.
</p>

  <p>
    علاوه بر دسته‌های اصلی که بررسی کردیم، پایتون چند عملگر خاص و بسیار کاربردی دیگر نیز دارد که برای کارهای مشخصی استفاده می‌شوند. در این اسلاید با دو دسته مهم از آن‌ها آشنا می‌شویم: <strong>عملگرهای عضویت</strong> و <strong>عملگرهای همانی</strong>.

  <hr style="border-top: 2px dashed #666; margin: 25px 0;">

  <h3 style="text-align: right; color: #007bff; font-weight: bold;">۱. عملگرهای عضویت (Membership Operators)</h3>

  <p>
    این عملگرها برای بررسی اینکه آیا یک مقدار خاص در یک دنباله (مانند یک لیست، یک تاپل یا یک رشته) وجود دارد یا خیر، استفاده می‌شوند. پاسخ این عملگرها همیشه <code>True</code> یا <code>False</code> است.
  </p>

  <p>
    <b>آنالوژی:</b> تصور کنید یک سبد میوه دارید. عملگر عضویت مانند این است که به سبد نگاه کنید و بپرسید: «آیا در این سبد "سیب" وجود دارد؟».
  </p>

  <h3 style="text-align: right; color: #28a745; font-weight: bold;">۲. عملگرهای همانی (Identity Operators)</h3>

  <p>
    این عملگرها دو متغیر را مقایسه می‌کنند تا ببینند آیا به <strong>یک شیء واحد در حافظه</strong> اشاره می‌کنند یا خیر. این مفهوم با مساوی بودن (<code>==</code>) که فقط مقادیر را مقایسه می‌کند، متفاوت است.
  </p>

  <p>
    <b>آنالوژی:</b> فرض کنید دو نفر (علی و رضا) هر دو یک پیراهن آبی یکسان خریده‌اند. مقدار آن‌ها (رنگ و مدل) یکی است (<code>==</code>)، اما دو پیراهن مجزا هستند (<code>is not</code>). حالا اگر به یک پیراهن خاص هم با نام «پیراهن علی» و هم با نام «لباس مهمانی» اشاره کنیم، این دو نام به یک شیء واحد اشاره دارند (<code>is</code>).
  </p>

  <table style="width: 100%; direction: rtl; border-collapse: collapse; margin-top: 15px;">
    <thead style="background-color: #f0f0f0;">
      <tr style="border: 2px solid #333;">

        <td style="padding: 12px; border: 2px solid #333;"><code>a is b</code></td>

        <td style="padding: 12px; border: 2px solid #333;"><code>a is not b</code></td>
      </tr>

  <pre style="background-color: #272822; color: #f8f8f2; padding: 15px; border-radius: 8px; direction: ltr; text-align: left; font-size: 15px;">
list1 = [1, 2, 3]
list2 = [1, 2, 3] # یک لیست جدید با مقادیر مشابه
list3 = list1     # list3 به همان شیء list1 اشاره می‌کند

print("list1 == list2:", list1 == list2)   # --> True (چون مقادیرشان برابر است)
print("list1 is list2:", list1 is list2)   # --> False (چون دو شیء مجزا در حافظه هستند)
print("list1 is list3:", list1 is list3)   # --> True (چون به یک شیء واحد اشاره دارند)
  </pre>

  <p style="background-color: #f9f9f9; padding: 12px; border-radius: 8px; font-weight: bold;">
    <strong>نکته کلیدی:</strong> از عملگر <code>==</code> برای بررسی <strong>برابری مقادیر</strong> و از عملگر <code>is</code> برای بررسی <strong>یکی بودن شیء در حافظه</strong> استفاده کنید. در اکثر مواقع شما با <code>==</code> کار خواهید کرد.
