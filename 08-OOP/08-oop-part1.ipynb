    <h4 style="color:#f56565;margin:20px 0 10px;font-size:1.3rem;">روش رویه‌ای (Procedural)</h4>
    <div style="background:#fff5f5;padding:15px;border-radius:10px;border-right:3px solid #f56565;margin-bottom:20px;color:#4a5568;">
        در این روش داده‌ها و توابع کاملاً جدا از هم هستند:

```python
# ❌ برنامه‌نویسی رویه‌ای - داده‌ها و توابع جدا

# داده‌ها (متغیرهای سرگردان!)
car1_color = "قرمز"
car1_speed = 0
car1_fuel = 50

car2_color = "آبی"
car2_speed = 0
car2_fuel = 40

# توابع جدا (باید همه داده‌ها را به عنوان پارامتر بگیرند)
def accelerate(speed, fuel):
    """تابع افزایش سرعت"""
    if fuel > 0:
        speed += 10
        fuel -= 1
        return speed, fuel
    return speed, fuel

def honk():
    """تابع بوق زدن"""
    print("بوق بوق!")

# استفاده از توابع
print("=== روش رویه‌ای ===")
print(f"خودرو 1 - رنگ: {car1_color}, سرعت: {car1_speed}, بنزین: {car1_fuel}")

# برای تغییر وضعیت باید همه چیز را دستی مدیریت کنیم
car1_speed, car1_fuel = accelerate(car1_speed, car1_fuel)
print(f"پس از شتاب - سرعت: {car1_speed}, بنزین: {car1_fuel}")
honk()

print("\n" + "="*50)
print("مشکلات این روش:")
print("• داده‌ها پراکنده هستند")
print("• باید همیشه متغیرها را دستی پاس دهیم")
print("• افزودن ویژگی جدید سخت است")
print("• خطاهای بیشتری ممکن است رخ دهد")
```

<div dir="rtl" style="text-align: right; margin: 2rem auto; max-width: 900px; font-family: 'Vazirmatn', sans-serif;">
    <h4 style="color: #48bb78; margin: 20px 0 10px 0; font-size: 1.3rem;">روش شیءگرا (Object-Oriented)</h4>
    <div style="background: #f0fff4; padding: 15px; border-radius: 10px; border-right: 3px solid #48bb78; margin-bottom: 20px;">
        <p style="color: #4a5568; margin-bottom: 10px;">در این روش همه چیز در یک بسته منظم قرار می‌گیرد:</p>
    </div>
</div>

```python
# ✅ برنامه‌نویسی شیءگرا - داده‌ها و رفتارها در یک بسته

class Car:
    """کلاس خودرو - تمام ویژگی‌ها و رفتارهای مرتبط در یکجا"""
    
    def __init__(self, color, initial_fuel=50):
        """سازنده - تنظیم وضعیت اولیه"""
        self.color = color          # ویژگی: رنگ خودرو
        self.speed = 0              # ویژگی: سرعت فعلی
        self.fuel = initial_fuel    # ویژگی: مقدار بنزین
        self.is_engine_on = False   # ویژگی: وضعیت موتور
    
    def start_engine(self):
        """متد: روشن کردن موتور"""
        if self.fuel > 0:
            self.is_engine_on = True
            print(f"🚗 موتور خودرو {self.color} روشن شد!")
        else:
            print("❌ بنزین کافی نیست!")
    
    def accelerate(self):
        """متد: افزایش سرعت"""
        if self.is_engine_on and self.fuel > 0:
            self.speed += 10
            self.fuel -= 1
            print(f"🚀 سرعت به {self.speed} کیلومتر افزایش یافت. بنزین: {self.fuel}L")
        else:
            print("⚠️ اول موتور را روشن کنید!")
    
    def honk(self):
        """متد: بوق زدن"""
        print(f"🚨 خودرو {self.color}: بوق بوق!")
    
    def get_status(self):
        """متد: دریافت وضعیت فعلی"""
        status = "روشن" if self.is_engine_on else "خاموش"
        return f"خودرو {self.color} - سرعت: {self.speed} - بنزین: {self.fuel} - موتور: {status}"

# استفاده از کلاس
print("=== روش شیءگرا ===")

# ساخت دو خودرو مختلف
car1 = Car("قرمز", 50)
car2 = Car("آبی", 40)

# کار با خودرو اول
print("\n--- خودرو اول ---")
print(car1.get_status())
car1.start_engine()
car1.accelerate()
car1.accelerate()
car1.honk()

# کار با خودرو دوم
print("\n--- خودرو دوم ---")
print(car2.get_status())
car2.honk()

print("\n" + "="*50)
print("مزایای این روش:")
print("• همه چیز سازمان‌یافته است")
print("• نیازی به پاس دادن دستی داده‌ها نیست")
print("• افزودن ویژگی جدید آسان است")
print("• خطاهای کمتری رخ می‌دهد")
print("• قابلیت استفاده مجدد دارد")
```

<h4 style="font-size: 1.4rem; margin-bottom: 20px; color: #2d3748; text-align: center;">کلاس (Class)</h4>
<div style="color: #4a5568; line-height: 1.8;">
<p><strong>تعریف:</strong> کلاس یک <strong>قالب</strong> یا <strong>طرح کلی</strong> است که مشخص می‌کند یک شیء چه ویژگی‌ها و رفتارهایی دارد.</p>

        <div style="color: #4a5568; line-height: 1.8;">
            <p><strong>تعریف:</strong> شیء یک <strong>نمونه واقعی</strong> از یک کلاس است که در حافظه ساخته می‌شود.</p>
            <div style="background: white; padding: 15px; border-radius: 10px; margin: 15px 0; border-right: 3px solid #48bb78; box-sizing: border-box;">
                <strong>مثال‌های واقعی از شیء:</strong>
                <ul style="padding-right: 20px; margin-top: 10px;">

بیایید یک کلاس Student ایجاد کنیم تا رابطه بین کلاس و شیء را به صورت عملی ببینیم. این کلاس نمایانگر یک دانشجو در سیستم دانشگاه است.

<strong>🎓 هدف:</strong> ایجاد سیستمی برای مدیریت اطلاعات دانشجویان شامل نام، شماره دانشجویی، نمرات و عملیات مرتبط

```python
# تعریف کلاس Student (طرح کلی)
class Student:
    """
    کلاس Student نمایانگر یک دانشجو در سیستم دانشگاه
    
    این کلاس مشخص می‌کند که هر دانشجو چه ویژگی‌هایی دارد و چه کارهایی می‌تواند انجام دهد.
    """
    
    def __init__(self, name, student_id, major="علوم کامپیوتر"):
        """
        سازنده کلاس - این متد هنگام ساخت هر شیء جدید به طور خودکار فراخوانی می‌شود
        
        پارامترها:
        name (str): نام دانشجو
        student_id (str): شماره دانشجویی
        major (str): رشته تحصیلی (پیش‌فرض: علوم کامپیوتر)
        """
        self.name = name              # ویژگی: نام دانشجو
        self.student_id = student_id  # ویژگی: شماره دانشجویی
        self.major = major            # ویژگی: رشته تحصیلی
        self.grades = []              # ویژگی: لیست نمرات (خالی در ابتدا)
        self.total_credits = 0        # ویژگی: مجموع واحدهای گذرانده
        
        print(f"✅ دانشجوی جدید ایجاد شد: {self.name} ({self.student_id})")
    
    def add_grade(self, grade, credit=1):
        """
        افزودن نمره جدید برای دانشجو
        
        پارامترها:
        grade (float): نمره (بین 0 تا 20)
        credit (int): تعداد واحد درس
        """
        if 0 <= grade <= 20:
            self.grades.append({
                'grade': grade,
                'credit': credit
            })
            self.total_credits += credit
            print(f"📝 نمره {grade} با {credit} واحد برای {self.name} ثبت شد.")
        else:
            print(f"❌ نمره {grade} معتبر نیست! باید بین 0 تا 20 باشد.")
    
    def calculate_average(self):
        """
        محاسبه میانگین وزنی نمرات
        
        بازگشت:
        float: میانگین وزنی نمرات
        """
        if not self.grades:
            print("⚠️ هیچ نمره‌ای ثبت نشده است!")
            return 0
        
        total_weighted = sum(grade['grade'] * grade['credit'] for grade in self.grades)
        average = total_weighted / self.total_credits
        
        print(f"📊 میانگین {self.name}: {average:.2f}")
        return average
    
    def get_status(self):
        """
        دریافت وضعیت کامل دانشجو
        
        بازگشت:
        str: رشته‌ای حاوی اطلاعات کامل دانشجو
        """
        status = f"""
🎓 وضعیت دانشجو:
├─ نام: {self.name}
├─ شماره دانشجویی: {self.student_id}
├─ رشته: {self.major}
├─ تعداد نمرات ثبت‌شده: {len(self.grades)}
├─ مجموع واحدها: {self.total_credits}
└─ میانگین: {self.calculate_average():.2f if self.grades else 'تعیین نشده'}
"""
        return status
    
    def is_passing(self, passing_grade=12):
        """
        بررسی وضعیت قبولی دانشجو
        
        پارامترها:
        passing_grade (float): نمره قبولی (پیش‌فرض: 12)
        
        بازگشت:
        bool: True اگر میانگین بالاتر از نمره قبولی باشد
        """
        average = self.calculate_average()
        if average >= passing_grade:
            print(f"🎉 {self.name} با میانگین {average:.2f} قبول است!")
            return True
        else:
            print(f"😞 {self.name} با میانگین {average:.2f} مردود است!")
            return False

# حالا از کلاس استفاده می‌کنیم - ساخت اشیاء (نمونه‌های واقعی)
print("=" * 60)
print("ساخت اشیاء از کلاس Student")
print("=" * 60)

# ساخت سه دانشجوی مختلف از یک کلاس
student1 = Student("علی احمدی", "4012345678", "مهندسی کامپیوتر")
student2 = Student("سارا محمدی", "4012345679", "ریاضی محض")
student3 = Student("رضا کریمی", "4012345680")  # رشته پیش‌فرض استفاده می‌شود

print("\n" + "=" * 60)
print("ثبت نمرات برای دانشجویان")
print("=" * 60)

# ثبت نمرات برای علی
student1.add_grade(18, 3)  # نمره 18 با 3 واحد
student1.add_grade(16, 4)  # نمره 16 با 4 واحد
student1.add_grade(19, 3)  # نمره 19 با 3 واحد

# ثبت نمرات برای سارا
student2.add_grade(14, 3)
student2.add_grade(15, 4)
student2.add_grade(11, 3)  # نمره پایین

# رضا هنوز نمره‌ای ندارد

print("\n" + "=" * 60)
print("بررسی وضعیت دانشجویان")
print("=" * 60)

# نمایش وضعیت همه دانشجویان
print("\n--- دانشجوی اول ---")
print(student1.get_status())
student1.is_passing()

print("\n--- دانشجوی دوم ---")
print(student2.get_status())
student2.is_passing()

print("\n--- دانشجوی سوم ---")
print(student3.get_status())
student3.is_passing()

print("\n" + "=" * 60)
print("نکات مهم این مثال:")
print("• از یک کلاس، چندین شیء مستقل ساخته شد")
print("• هر شیء داده‌های مخصوص به خودش را دارد")
print("• تغییر در یک شیء بر شیء دیگر تأثیر نمی‌گذارد")
print("• همه اشیاء از یک ساختار پیروی می‌کنند")
print("=" * 60)
```

```python
class BankAccount:
    """
    کلاس BankAccount نمایانگر یک حساب بانکی
    
    این کلاس تمام عملیات مربوط به یک حساب بانکی را مدیریت می‌کند.
    """
    
    # ویژگی کلاس - بین همه حساب‌ها مشترک است
    bank_name = "بانک ملی ایران"
    total_accounts = 0
    
    def __init__(self, account_number, owner_name, initial_balance=0):
        """
        سازنده کلاس - ایجاد یک حساب بانکی جدید
        
        پارامترها:
        account_number (str): شماره حساب
        owner_name (str): نام صاحب حساب
        initial_balance (float): موجودی اولیه (پیش‌فرض: 0)
        """
        # ویژگی‌های نمونه - برای هر حساب منحصر به فرد
        self.account_number = account_number    # ویژگی: شماره حساب
        self.owner_name = owner_name            # ویژگی: نام صاحب حساب
        self.balance = initial_balance          # ویژگی: موجودی حساب
        self.transaction_history = []           # ویژگی: تاریخچه تراکنش‌ها
        self.is_active = True                   # ویژگی: وضعیت فعال بودن حساب
        
        # به‌روزرسانی ویژگی کلاس
        BankAccount.total_accounts += 1
        
        # ثبت اولین تراکنش (افتتاح حساب)
        self._add_transaction("افتتاح حساب", initial_balance, initial_balance)
        
        print(f"✅ حساب جدید افتتاح شد: {self.owner_name} - شماره حساب: {self.account_number}")
    
    def _add_transaction(self, transaction_type, amount, new_balance):
        """
        متد خصوصی: افزودن تراکنش به تاریخچه
        (متدهایی که با _ شروع می‌شوند معمولاً خصوصی در نظر گرفته می‌شوند)
        """
        from datetime import datetime
        transaction = {
            'type': transaction_type,
            'amount': amount,
            'balance': new_balance,
            'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        self.transaction_history.append(transaction)
    
    def deposit(self, amount):
        """
        واریز پول به حساب
        
        پارامترها:
        amount (float): مبلغ واریزی
        
        بازگشت:
        bool: True اگر واریز موفقیت‌آمیز بود
        """
        if not self.is_active:
            print("❌ حساب غیرفعال است! نمی‌توان واریز انجام داد.")
            return False
        
        if amount > 0:
            self.balance += amount
            self._add_transaction("واریز", amount, self.balance)
            print(f"✅ {amount:,} تومان واریز شد. موجودی جدید: {self.balance:,} تومان")
            return True
        else:
            print("❌ مبلغ واریز باید بیشتر از صفر باشد!")
            return False
    
    def withdraw(self, amount):
        """
        برداشت پول از حساب
        
        پارامترها:
        amount (float): مبلغ برداشت
        
        بازگشت:
        bool: True اگر برداشت موفقیت‌آمیز بود
        """
        if not self.is_active:
            print("❌ حساب غیرفعال است! نمی‌توان برداشت انجام داد.")
            return False
        
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                self._add_transaction("برداشت", -amount, self.balance)
                print(f"✅ {amount:,} تومان برداشت شد. موجودی جدید: {self.balance:,} تومان")
                return True
            else:
                print(f"❌ موجودی کافی نیست! موجودی فعلی: {self.balance:,} تومان")
                return False
        else:
            print("❌ مبلغ برداشت باید بیشتر از صفر باشد!")
            return False
    
    def transfer(self, amount, target_account):
        """
        انتقال پول به حساب دیگر
        
        پارامترها:
        amount (float): مبلغ انتقال
        target_account (BankAccount): حساب مقصد
        
        بازگشت:
        bool: True اگر انتقال موفقیت‌آمیز بود
        """
        if not self.is_active or not target_account.is_active:
            print("❌ یکی از حساب‌ها غیرفعال است!")
            return False
        
        print(f"\n💸 شروع انتقال {amount:,} تومان از {self.owner_name} به {target_account.owner_name}")
        
        # برداشت از حساب مبدأ
        if self.withdraw(amount):
            # واریز به حساب مقصد
            if target_account.deposit(amount):
                self._add_transaction(f"انتقال به {target_account.account_number}", -amount, self.balance)
                target_account._add_transaction(f"انتقال از {self.account_number}", amount, target_account.balance)
                print(f"✅ انتقال با موفقیت انجام شد!")
                return True
            else:
                # اگر واریز به مقصد失败 شد، پول را برمی‌گردانیم
                self.deposit(amount)
                print("❌ انتقال ناموفق! پول به حساب برگشت.")
                return False
        return False
    
    def get_balance(self):
        """
        دریافت موجودی حساب
        
        بازگشت:
        float: موجودی فعلی
        """
        return self.balance
    
    def get_statement(self):
        """
        دریافت صورت حساب
        
        بازگشت:
        str: صورت حساب فرمت‌بندی شده
        """
        statement = f"""
🏦 {BankAccount.bank_name}
┌─ صورت حساب
├─ صاحب حساب: {self.owner_name}
├─ شماره حساب: {self.account_number}
├─ موجودی فعلی: {self.balance:,} تومان
├─ وضعیت: {'فعال' if self.is_active else 'غیرفعال'}
└─ تاریخچه تراکنش‌ها:
"""
        
        for i, transaction in enumerate(self.transaction_history[-5:], 1):  # فقط 5 تراکنش آخر
            amount_str = f"+{transaction['amount']:,}" if transaction['amount'] > 0 else f"{transaction['amount']:,}"
            statement += f"\n   {i}. {transaction['timestamp']} - {transaction['type']}: {amount_str} - موجودی: {transaction['balance']:,}"
        
        return statement
    
    def deactivate_account(self):
        """غیرفعال کردن حساب"""
        if self.balance == 0:
            self.is_active = False
            print("✅ حساب غیرفعال شد.")
            return True
        else:
            print("❌ برای غیرفعال کردن حساب، موجودی باید صفر باشد!")
            return False
    
    @classmethod
    def get_bank_info(cls):
        """
        متد کلاس: دریافت اطلاعات بانک
        """
        return f"بانک: {cls.bank_name} - تعداد کل حساب‌ها: {cls.total_accounts}"

# استفاده از سیستم حساب بانکی
print("=" * 70)
print("سیستم مدیریت حساب بانکی")
print("=" * 70)

# اطلاعات کلی بانک
print(BankAccount.get_bank_info())

# ایجاد حساب‌های مختلف
print("\n" + "-" * 70)
print("افتتاح حساب‌های جدید")
print("-" * 70)

account1 = BankAccount("1234567890", "علی احمدی", 1_000_000)  # 1 میلیون تومان
account2 = BankAccount("1234567891", "سارا محمدی", 500_000)   # 500 هزار تومان
account3 = BankAccount("1234567892", "رضا کریمی")             # بدون موجودی اولیه

print("\n" + "-" * 70)
print("انجام تراکنش‌ها")
print("-" * 70)

# واریزها
account1.deposit(200_000)
account2.deposit(100_000)
account3.deposit(50_000)

print("\n" + "-" * 70)
print("برداشت‌ها")
print("-" * 70)

account1.withdraw(50_000)
account2.withdraw(600_000)  # این باید失败 شود (موجودی ناکافی)

print("\n" + "-" * 70)
print("انتقال وجه")
print("-" * 70)

account1.transfer(100_000, account3)

print("\n" + "-" * 70)
print("صورت حساب‌ها")
print("-" * 70)

print("\n--- حساب اول ---")
print(account1.get_statement())

print("\n--- حساب دوم ---")
print(account2.get_statement())

print("\n--- حساب سوم ---")
print(account3.get_statement())

print("\n" + "=" * 70)
print("اطلاعات نهایی بانک")
print("=" * 70)
print(BankAccount.get_bank_info())

print("\n" + "=" * 70)
print("نکات آموزشی این مثال:")
print("=" * 70)
print("• استفاده از ویژگی‌های کلاس (bank_name, total_accounts)")
print("• استفاده از ویژگی‌های نمونه (balance, transaction_history)")
print("• متدهای معمولی (deposit, withdraw, transfer)")
print("• متد خصوصی (_add_transaction)")
print("• متد کلاس (get_bank_info)")
print("• مدیریت وضعیت (is_active)")
print("• ثبت تاریخچه تراکنش‌ها")
print("• کنترل خطا و اعتبارسنجی")
print("=" * 70)
```

<div style="background:linear-gradient(135deg,#667eea10 0%,#f093fb10 100%);border-right:4px solid #667eea;border-radius:15px;padding:25px;margin:20px auto;line-height:1.8;box-sizing:border-box;">
<p style="font-size:1.05rem;color:#4a5568;">
متد __init__ یک متد جادویی (Magic Method) و بسیار ویژه در پایتون است که به آن <strong>سازنده (Constructor)</strong> می‌گویند. این متد به محض اینکه یک شیء جدید از روی کلاس ساخته می‌شود، به صورت <strong>خودکار</strong> فراخوانی می‌شود.
</p>
<p style="font-size:1.05rem;color:#4a5568;margin-top:10px;">
<strong>وظیفه اصلی آن</strong>، مقداردهی اولیه به ویژگی‌های (Attributes) شیء است. یعنی داده‌های اولیه و ضروری برای شروع به کار شیء را تنظیم می‌کند.
</p>
<!-- کادر نکته -->
<div style="background:#e6fffa;padding:15px;border-radius:10px;margin:20px 0;border-right:3px solid #38b2ac;box-sizing:border-box;">
<strong>🚗 مثال دنیای واقعی:</strong> وقتی یک ماشین در کارخانه تولید می‌شود، همان لحظه اول رنگ، مدل و شماره شاسی آن مشخص می‌شود. متد __init__ دقیقاً همین نقش را برای یک شیء ایفا می‌کند و هویت اولیه آن را تعیین می‌کند.
</div>
<!-- ستون‌های ویژگی‌ها -->
<div style="display:grid;grid-template-columns:1fr 1fr;gap:15px;margin:20px 0;">
<div style="background:white;padding:15px;border-radius:10px;border:1px solid #e2e8f0;box-sizing:border-box;">
<h5 style="color:#2d3748;margin-bottom:10px;">📋 ساختار اصلی:</h5>
<ul style="color:#4a5568;padding-right:15px;margin:0;">
<li>با دو آندرلاین شروع و تمام می‌شود.</li>
<li>اولین پارامتر آن همیشه <code>self</code> است.</li>
<li>پارامترهای بعدی، ورودی‌های لازم برای ساخت شیء هستند.</li>
<li>هیچ مقداری را <code>return</code> نمی‌کند.</li>

<!-- کانتینر اصلی -->
<div style="background:linear-gradient(135deg,#f6ad5510 0%,#ed893610 100%);border-right:4px solid #ed8936;border-radius:15px;padding:25px;margin:20px auto;line-height:1.8;box-sizing:border-box;">
<p style="font-size:1.05rem;color:#4a5568;">
self یک قرارداد در پایتون است که به <strong>نمونه فعلی (instance)</strong> کلاس اشاره دارد. به زبان ساده، self همان شیئی است که متد روی آن فراخوانی شده است. از طریق self، یک متد می‌تواند به سایر ویژگی‌ها و متدهای همان شیء دسترسی پیدا کند.
</p>
<p style="font-size:1.05rem;color:#4a5568;margin-top:10px;">
وقتی شما متدی مانند my_object.my_method(arg1) را فراخوانی می‌کنید، پایتون به صورت خودکار آن را به MyClass.my_method(my_object, arg1) تبدیل می‌کند. پارامتر self همان my_object است.
</p>
<!-- کادر نکته -->
<div style="background:#fffaf0;padding:15px;border-radius:10px;margin:20px 0;border-right:3px solid #f6ad55;box-sizing:border-box;">
<strong>🔑 مثال کلیدی:</strong> تصور کنید چند نفر به نام "علی" در یک اتاق هستند. اگر یکی از آن‌ها بگوید "<strong>من</strong> تشنه هستم"، کلمه "من" مشخص می‌کند که کدام علی صحبت می‌کند. self در برنامه‌نویسی شیءگرا دقیقاً نقش همین "من" را برای اشیاء ایفا می‌کند.
</div>

<h4 style="color: #2d3748; margin-bottom: 15px;">🎯 قانون طلایی self</h4>
<p style="color: #4a5568; line-height: 1.8;">
هرگاه در داخل یک متد نیاز به <strong>خواندن یا نوشتن</strong> یک ویژگی (Attribute) یا فراخوانی یک متد دیگر <strong>متعلق به همان شیء</strong> را داشتید، <strong>باید</strong> از self استفاده کنید.
</p>
<p style="color: #4a5568; line-height: 1.8; margin-top: 10px;">
• برای دسترسی به ویژگی: <code>self.my_attribute</code><br>
• برای فراخوانی متد: <code>self.my_other_method()</code>

```python
import datetime
import random

class BankAccount:
    """
    این کلاس یک حساب بانکی را با ویژگی‌ها و عملیات‌های مختلف شبیه‌سازی می‌کند.
    """
    def __init__(self, owner_name, initial_balance=0):
        # 1. اعتبارسنجی ورودی‌ها
        if initial_balance < 0:
            raise ValueError("موجودی اولیه نمی‌تواند منفی باشد!")
        if len(owner_name) < 3:
            raise ValueError("نام صاحب حساب باید حداقل 3 کاراکتر باشد.")

        # 2. مقداردهی ویژگی‌ها از پارامترهای ورودی
        self.owner = owner_name
        self.balance = initial_balance
        
        # 3. ایجاد مقادیر پیش‌فرض و محاسبه‌شده
        self.account_number = random.randint(100000, 999999)
        self.creation_date = datetime.date.today()
        self.is_active = True
        self.transactions = []

        # 4. فراخوانی یک متد دیگر در زمان ساخت
        self._log_transaction(f"حساب در تاریخ {self.creation_date} ایجاد شد.")
        print(f"حساب جدید برای '{self.owner}' با شماره {self.account_number} با موفقیت ایجاد شد.")

    def _log_transaction(self, description):
        """یک متد داخلی برای ثبت تراکنش‌ها (برای استفاده در داخل کلاس)"""
        self.transactions.append(f"[{datetime.datetime.now()}] - {description}")

# --- ساختن اشیاء از روی کلاس ---

# ساخت یک حساب معتبر
try:
    account1 = BankAccount("علی رضایی", 50000)
    print(f"موجودی اولیه حساب علی: {account1.balance}")
except ValueError as e:
    print(f"خطا در ساخت حساب: {e}")

print("-" * 20)

# تلاش برای ساخت یک حساب نامعتبر با موجودی منفی
try:
    account2 = BankAccount("سارا محمدی", -1000)
except ValueError as e:
    print(f"خطا در ساخت حساب سارا: {e}")
```

<p style="font-size:1.05rem;color:#4a5568;">
در کد زیر، ما یک کلاس ساده Player برای یک بازی تعریف می‌کنیم. هر بازیکن (شیء) نام و سطح سلامتی (health) منحصر به فرد خود را دارد.
</p>
<!-- کادر نکته -->
<div style="background:#fffaf0;padding:15px;border-radius:10px;margin:20px 0;border-right:3px solid #f6ad55;box-sizing:border-box;">
<strong>🎯 در این کد مشاهده خواهید کرد:</strong>
<p style="margin:10px 0 0 0;">
وقتی متد <code>take_damage</code> را روی یکی از بازیکنان فراخوانی می‌کنیم، self تضمین می‌کند که <strong>فقط و فقط سطح سلامتی همان بازیکن</strong> تغییر کند. self به پایتون می‌گوید که کدام شیء هدف این تغییر است و از تأثیر آن بر سایر اشیاء جلوگیری می‌کند.
</p>
</div>
</div>
</div>

```python
class Player:
    """
    این کلاس یک بازیکن در یک بازی را نشان می‌دهد.
    """
    def __init__(self, name):
        # self در اینجا به شیء جدیدی که در حال ساخت است اشاره دارد
        self.name = name
        self.health = 100  # همه بازیکنان با سلامتی 100 شروع می‌کنند

    def take_damage(self, amount):
        """از سطح سلامتی بازیکن کم می‌کند."""
        # self در اینجا به شیئی اشاره دارد که این متد روی آن فراخوانی شده
        # مثلاً player1 یا player2
        self.health -= amount
        print(f"{self.name} مقدار {amount} آسیب دید! سلامتی باقی‌مانده: {self.health}")

    def show_status(self):
        """وضعیت فعلی بازیکن را نمایش می‌دهد."""
        # self به ما امکان دسترسی به نام و سلامتی شیء فعلی را می‌دهد
        print(f"بازیکن: {self.name}, سطح سلامتی: {self.health}")

# --- ساختن دو شیء مستقل از کلاس Player ---
player1 = Player("آریا")
player2 = Player("سارا")

# نمایش وضعیت اولیه هر دو بازیکن
print("--- وضعیت اولیه ---")
player1.show_status() # در اینجا self همان player1 است
player2.show_status() # در اینجا self همان player2 است

print("\n--- بازیکن 1 آسیب می‌بیند ---")
# متد take_damage را فقط روی player1 فراخوانی می‌کنیم
player1.take_damage(30) # self در داخل متد به player1 اشاره می‌کند

# نمایش وضعیت نهایی هر دو بازیکن
print("\n--- وضعیت نهایی ---")
player1.show_status() # می‌بینیم که سلامتی player1 کم شده
player2.show_status() # اما سلامتی player2 بدون تغییر باقی مانده است
```
