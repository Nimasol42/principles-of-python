        ساختار پایه وراثت
    </h3>
    <div style="background: linear-gradient(135deg, #4299e110 0%, #3182ce10 100%); border-right: 4px solid #4299e1; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        در پایتون، برای ایجاد وراثت کافیست نام کلاس والد را در پرانتز بعد از نام کلاس فرزند بنویسیم. از تابع <code>super()</code> برای دسترسی به متدها و ویژگی‌های کلاس والد استفاده می‌کنیم.
    </div>
</div>

```python
# کلاس والد (Parent/Base Class)
class ParentClass:
    def __init__(self):
        self.parent_attribute = "مقدار والد"
    
    def parent_method(self):
        print("این متد از کلاس والد است")

# کلاس فرزند (Child/Derived Class)
class ChildClass(ParentClass):
    def __init__(self):
        super().__init__()  # فراخوانی سازنده والد
        self.child_attribute = "مقدار فرزند"
    
    def child_method(self):
        print("این متد مخصوص کلاس فرزند است")

# استفاده
child = ChildClass()
child.parent_method()  # دسترسی به متد والد
child.child_method()   # دسترسی به متد فرزند
print(child.parent_attribute)  # دسترسی به ویژگی والد
print(child.child_attribute)   # دسترسی به ویژگی فرزند
```

<div dir="rtl" style="text-align: right; margin: 2rem auto; max-width: 900px; font-family: 'Vazirmatn', sans-serif; box-sizing: border-box;">
    <h3 style="font-size: 1.5rem; font-weight: 600; color: #4a5568; margin: 25px 0 15px 0; padding-right: 20px; border-right: 6px solid #667eea;">
        کلاس والد: Animal
    </h3>
    <div style="background: linear-gradient(135deg, #4299e110 0%, #3182ce10 100%); border-right: 4px solid #4299e1; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        این کلاس پایه شامل ویژگی‌ها و رفتارهای مشترک همه حیوانات است. تمام حیوانات نام و سن دارند و می‌توانند غذا بخورند، بخوابند و صدا تولید کنند.
    </div>
</div>

```python
class Animal:
    """کلاس پایه برای همه حیوانات"""
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def eat(self):
        print(f"{self.name} در حال غذا خوردن است")
    
    def sleep(self):
        print(f"{self.name} در حال خوابیدن است")
    
    def make_sound(self):
        print("حیوان صدا می‌کند")
    
    def get_info(self):
        """نمایش اطلاعات پایه حیوان"""
        return f"نام: {self.name}, سن: {self.age} سال"
```

<div dir="rtl" style="text-align: right; margin: 2rem auto; max-width: 900px; font-family: 'Vazirmatn', sans-serif; box-sizing: border-box;">
    <h3 style="font-size: 1.5rem; font-weight: 600; color: #4a5568; margin: 25px 0 15px 0; padding-right: 20px; border-right: 6px solid #667eea;">
        کلاس‌های فرزند
    </h3>
    <div style="background: linear-gradient(135deg, #4299e110 0%, #3182ce10 100%); border-right: 4px solid #4299e1; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        هر کلاس فرزند ویژگی‌ها و رفتارهای خاص خود را دارد در حالی که ویژگی‌های مشترک را از کلاس والد به ارث می‌برد.
    </div>
</div>

```python
class Dog(Animal):
    """کلاس سگ که از Animal ارث‌بری می‌کند"""
    
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # فراخوانی سازنده والد
        self.breed = breed  # ویژگی اختصاصی سگ
    
    def make_sound(self):
        """بازنویسی متد والد - سگ پارس می‌کند"""
        print(f"{self.name} پارس می‌کند: واق واق!")
    
    def fetch(self):
        """متد اختصاصی سگ"""
        print(f"{self.name} توپ را می‌آورد")
    
    def get_info(self):
        """بازنویسی متد برای نمایش اطلاعات کامل سگ"""
        base_info = super().get_info()
        return f"{base_info}, نژاد: {self.breed}"


class Cat(Animal):
    """کلاس گربه که از Animal ارث‌بری می‌کند"""
    
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color  # ویژگی اختصاصی گربه
    
    def make_sound(self):
        """بازنویسی متد والد - گربه میو می‌کند"""
        print(f"{self.name} میو می‌کند: میو میو!")
    
    def scratch(self):
        """متد اختصاصی گربه"""
        print(f"{self.name} چنگ می‌زند")
    
    def get_info(self):
        """بازنویسی متد برای نمایش اطلاعات کامل گربه"""
        base_info = super().get_info()
        return f"{base_info}, رنگ: {self.color}"


class Bird(Animal):
    """کلاس پرنده که از Animal ارث‌بری می‌کند"""
    
    def __init__(self, name, age, can_fly):
        super().__init__(name, age)
        self.can_fly = can_fly  # ویژگی اختصاصی پرنده
    
    def make_sound(self):
        """بازنویسی متد والد - پرنده جیک جیک می‌کند"""
        print(f"{self.name} جیک جیک می‌کند!")
    
    def fly(self):
        """متد اختصاصی پرنده"""
        if self.can_fly:
            print(f"{self.name} در حال پرواز است")
        else:
            print(f"{self.name} نمی‌تواند پرواز کند")
    
    def get_info(self):
        """بازنویسی متد برای نمایش اطلاعات کامل پرنده"""
        base_info = super().get_info()
        flight_status = "قادر به پرواز" if self.can_fly else "ناتوان در پرواز"
        return f"{base_info}, وضعیت پرواز: {flight_status}"
```

<div dir="rtl" style="text-align: right; margin: 2rem auto; max-width: 900px; font-family: 'Vazirmatn', sans-serif; box-sizing: border-box;">
    <h3 style="font-size: 1.5rem; font-weight: 600; color: #4a5568; margin: 25px 0 15px 0; padding-right: 20px; border-right: 6px solid #667eea;">
        استفاده از کلاس‌ها
    </h3>
    <div style="background: linear-gradient(135deg, #4299e110 0%, #3182ce10 100%); border-right: 4px solid #4299e1; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        حالا می‌توانیم از کلاس‌های ایجاد شده استفاده کنیم و قدرت وراثت را در عمل ببینیم.
    </div>
</div>

```python
# ساخت اشیاء از کلاس‌های مختلف
dog = Dog("رکس", 3, "ژرمن شپرد")
cat = Cat("میشی", 2, "سفید")
bird = Bird("توتی", 1, True)

print("=" * 50)
print("🐾 اطلاعات حیوانات")
print("=" * 50)

# نمایش اطلاعات حیوانات
print(dog.get_info())
print(cat.get_info())
print(bird.get_info())

print("\n" + "=" * 50)
print("🎭 رفتارهای مشترک")
print("=" * 50)

# استفاده از متدهای مشترک (از والد)
dog.eat()    # رکس در حال غذا خوردن است
cat.sleep()  # میشی در حال خوابیدن است
bird.eat()   # توتی در حال غذا خوردن است

print("\n" + "=" * 50)
print("🎯 رفتارهای اختصاصی")
print("=" * 50)

# استفاده از متدهای اختصاصی
dog.fetch()      # رکس توپ را می‌آورد
cat.scratch()    # میشی چنگ می‌زند
bird.fly()       # توتی در حال پرواز است

print("\n" + "=" * 50)
print("🔊 صداهای حیوانات")
print("=" * 50)

# استفاده از متد بازنویسی شده
dog.make_sound()   # رکس پارس می‌کند: واق واق!
cat.make_sound()   # میشی میو می‌کند: میو میو!
bird.make_sound()  # توتی جیک جیک می‌کند!
```

```python
class Shape:
    """کلاس پایه برای اشکال هندسی"""
    
    def area(self):
        """محاسبه مساحت - باید در کلاس فرزند پیاده‌سازی شود"""
        raise NotImplementedError("متد area باید در کلاس فرزند پیاده‌سازی شود")
    
    def perimeter(self):
        """محاسبه محیط - باید در کلاس فرزند پیاده‌سازی شود"""
        raise NotImplementedError("متد perimeter باید در کلاس فرزند پیاده‌سازی شود")
    
    def get_info(self):
        """نمایش اطلاعات شکل"""
        return f"مساحت: {self.area():.2f}, محیط: {self.perimeter():.2f}"


class Rectangle(Shape):
    """کلاس مستطیل"""
    
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        """محاسبه مساحت مستطیل"""
        return self.width * self.height
    
    def perimeter(self):
        """محاسبه محیط مستطیل"""
        return 2 * (self.width + self.height)
    
    def get_info(self):
        """نمایش اطلاعات مستطیل"""
        base_info = super().get_info()
        return f"مستطیل - عرض: {self.width}, ارتفاع: {self.height} | {base_info}"


class Circle(Shape):
    """کلاس دایره"""
    
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        """محاسبه مساحت دایره"""
        return 3.14159 * self.radius ** 2
    
    def perimeter(self):
        """محاسبه محیط دایره"""
        return 2 * 3.14159 * self.radius
    
    def get_info(self):
        """نمایش اطلاعات دایره"""
        base_info = super().get_info()
        return f"دایره - شعاع: {self.radius} | {base_info}"


class Triangle(Shape):
    """کلاس مثلث"""
    
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    
    def area(self):
        """محاسبه مساحت مثلث با فرمول هرون"""
        s = (self.a + self.b + self.c) / 2
        return (s * (s - self.a) * (s - self.b) * (s - self.c)) ** 0.5
    
    def perimeter(self):
        """محاسبه محیط مثلث"""
        return self.a + self.b + self.c
    
    def get_info(self):
        """نمایش اطلاعات مثلث"""
        base_info = super().get_info()
        return f"مثلث - اضلاع: {self.a}, {self.b}, {self.c} | {base_info}"


# چندریختی در عمل
print("🔷 چندریختی در عمل: اشکال هندسی")
print("=" * 60)

shapes = [
    Rectangle(5, 10),
    Circle(7),
    Triangle(3, 4, 5)
]

for i, shape in enumerate(shapes, 1):
    print(f"{i}. {shape.get_info()}")

print("\n" + "=" * 60)
print("📊 محاسبات جداگانه:")
print("=" * 60)

# همچنین می‌توانیم به طور جداگانه با هر شکل کار کنیم
for shape in shapes:
    print(f"مساحت: {shape.area():.2f}")
    print(f"محیط: {shape.perimeter():.2f}")
    print("-" * 30)
```

    یک سیستم پرداخت آنلاین می‌سازیم که از روش‌های مختلف پرداخت پشتیبانی می‌کند. این مثال به خوبی قدرت وراثت و چندریختی را در یک سناریوی واقعی نشان می‌دهد.

```python
class PaymentMethod:
    """کلاس پایه برای روش‌های پرداخت"""
    
    def __init__(self, amount):
        self.amount = amount
        self.is_successful = False
        self.transaction_id = None
    
    def process_payment(self):
        """این متد باید در کلاس‌های فرزند بازنویسی شود"""
        raise NotImplementedError("متد process_payment باید در کلاس فرزند پیاده‌سازی شود")
    
    def validate_amount(self):
        """اعتبارسنجی مقدار پرداخت - مشترک بین همه روش‌ها"""
        if self.amount <= 0:
            raise ValueError("مبلغ پرداخت باید بزرگتر از صفر باشد")
        return True
    
    def generate_transaction_id(self):
        """تولید شناسه تراکنش - مشترک بین همه روش‌ها"""
        import uuid
        self.transaction_id = str(uuid.uuid4())[:8].upper()
        return self.transaction_id
    
    def send_receipt(self):
        """ارسال رسید پرداخت"""
        if self.is_successful and self.transaction_id:
            print(f"📧 رسید پرداخت {self.amount:,} تومان ارسال شد")
            print(f"   شناسه تراکنش: {self.transaction_id}")
        else:
            print("❌ امکان ارسال رسید وجود ندارد - تراکنش ناموفق")
    
    def get_payment_info(self):
        """نمایش اطلاعات پرداخت"""
        status = "✅ موفق" if self.is_successful else "❌ ناموفق"
        return f"مبلغ: {self.amount:,} تومان | وضعیت: {status}"
```

```python
class CreditCard(PaymentMethod):
    """پرداخت با کارت بانکی"""
    
    def __init__(self, amount, card_number, cvv, card_holder):
        super().__init__(amount)
        self.card_number = card_number
        self.cvv = cvv
        self.card_holder = card_holder
    
    def validate_card(self):
        """اعتبارسنجی اطلاعات کارت"""
        if len(self.card_number) != 16 or not self.card_number.isdigit():
            raise ValueError("شماره کارت باید 16 رقمی باشد")
        if len(self.cvv) != 3 or not self.cvv.isdigit():
            raise ValueError("CVV باید 3 رقمی باشد")
        return True
    
    def process_payment(self):
        """پردازش پرداخت با کارت بانکی"""
        try:
            self.validate_amount()
            self.validate_card()
            
            print(f"💳 در حال پردازش پرداخت {self.amount:,} تومان با کارت...")
            print(f"   صاحب کارت: {self.card_holder}")
            print(f"   شماره کارت: ****{self.card_number[-4:]}")
            
            # شبیه‌سازی پردازش پرداخت
            self.generate_transaction_id()
            
            # در دنیای واقعی اینجا با درگاه بانکی ارتباط برقرار می‌شود
            self.is_successful = True
            print("✅ پرداخت با کارت موفق بود")
            
        except Exception as e:
            print(f"❌ خطا در پرداخت با کارت: {e}")
            self.is_successful = False
        
        return self.is_successful
    
    def get_payment_info(self):
        """نمایش اطلاعات پرداخت کارتی"""
        base_info = super().get_payment_info()
        return f"کارت بانکی - {base_info}"


class DigitalWallet(PaymentMethod):
    """پرداخت با کیف پول الکترونیکی"""
    
    def __init__(self, amount, wallet_id, wallet_balance, wallet_type):
        super().__init__(amount)
        self.wallet_id = wallet_id
        self.wallet_balance = wallet_balance
        self.wallet_type = wallet_type  # مثلاً: زرین‌پال، پی‌پال
    
    def process_payment(self):
        """پردازش پرداخت با کیف پول"""
        try:
            self.validate_amount()
            
            print(f"👛 در حال پردازش پرداخت {self.amount:,} تومان با کیف پول...")
            print(f"   نوع کیف پول: {self.wallet_type}")
            print(f"   شناسه کیف پول: {self.wallet_id}")
            print(f"   موجودی فعلی: {self.wallet_balance:,} تومان")
            
            if self.wallet_balance >= self.amount:
                self.wallet_balance -= self.amount
                self.generate_transaction_id()
                self.is_successful = True
                print(f"✅ پرداخت موفق - موجودی جدید: {self.wallet_balance:,} تومان")
            else:
                print("❌ موجودی کافی نیست")
                self.is_successful = False
            
        except Exception as e:
            print(f"❌ خطا در پرداخت با کیف پول: {e}")
            self.is_successful = False
        
        return self.is_successful
    
    def get_payment_info(self):
        """نمایش اطلاعات پرداخت کیف پولی"""
        base_info = super().get_payment_info()
        return f"کیف پول {self.wallet_type} - {base_info}"


class CryptoPayment(PaymentMethod):
    """پرداخت با رمزارز"""
    
    def __init__(self, amount, crypto_type, wallet_address):
        super().__init__(amount)
        self.crypto_type = crypto_type  # بیت‌کوین، اتریوم و...
        self.wallet_address = wallet_address
        self.exchange_rate = self.get_exchange_rate()
    
    def get_exchange_rate(self):
        """نرخ تبدیل رمزارز به تومان (شبیه‌سازی)"""
        rates = {
            "Bitcoin": 2_500_000_000,
            "Ethereum": 150_000_000,
            "Tether": 50_000
        }
        return rates.get(self.crypto_type, 50_000)
    
    def calculate_crypto_amount(self):
        """محاسبه مقدار معادل رمزارز"""
        return self.amount / self.exchange_rate
    
    def process_payment(self):
        """پردازش پرداخت با رمزارز"""
        try:
            self.validate_amount()
            
            crypto_amount = self.calculate_crypto_amount()
            
            print(f"₿ در حال پردازش پرداخت {self.amount:,} تومان با رمزارز...")
            print(f"   نوع رمزارز: {self.crypto_type}")
            print(f"   آدرس کیف پول: {self.wallet_address[:10]}...{self.wallet_address[-10:]}")
            print(f"   مقدار معادل: {crypto_amount:.8f} {self.crypto_type}")
            
            # شبیه‌سازی تأیید تراکنش بلاکچین
            self.generate_transaction_id()
            self.is_successful = True
            print("✅ پرداخت با رمزارز موفق بود")
            
        except Exception as e:
            print(f"❌ خطا در پرداخت با رمزارز: {e}")
            self.is_successful = False
        
        return self.is_successful
    
    def get_payment_info(self):
        """نمایش اطلاعات پرداخت رمزارزی"""
        base_info = super().get_payment_info()
        crypto_amount = self.calculate_crypto_amount()
        return f"{self.crypto_type} - {base_info} | معادل: {crypto_amount:.6f} {self.crypto_type}"
```

```python
class PaymentProcessor:
    """کلاس پردازش‌گر پرداخت - نمایش چندریختی"""
    
    def __init__(self):
        self.payment_history = []
    
    def execute_payment(self, payment_method):
        """اجرای پرداخت - این متد با همه انواع پرداخت کار می‌کند"""
        print("\n" + "="*60)
        print("🔄 شروع پردازش پرداخت...")
        print("="*60)
        
        # اینجاست که چندریختی اتفاق می‌افتد!
        # ما نمی‌دانیم payment_method دقیقاً از چه نوعی است
        # اما مطمئنیم که متد process_payment را دارد
        success = payment_method.process_payment()
        
        if success:
            payment_method.send_receipt()
            print("✅ تراکنش با موفقیت انجام شد")
        else:
            print("❌ تراکنش ناموفق بود")
        
        # ذخیره در تاریخچه
        self.payment_history.append({
            'method': type(payment_method).__name__,
            'amount': payment_method.amount,
            'success': success,
            'transaction_id': payment_method.transaction_id
        })
        
        print("="*60)
        return success
    
    def show_payment_history(self):
        """نمایش تاریخچه پرداخت‌ها"""
        print("\n" + "="*60)
        print("📋 تاریخچه پرداخت‌ها")
        print("="*60)
        
        if not self.payment_history:
            print("هیچ تراکنشی ثبت نشده است")
            return
        
        for i, payment in enumerate(self.payment_history, 1):
            status = "✅ موفق" if payment['success'] else "❌ ناموفق"
            print(f"{i}. {payment['method']} - {payment['amount']:,} تومان - {status}")
            if payment['transaction_id']:
                print(f"   شناسه تراکنش: {payment['transaction_id']}")
        
        print("="*60)
```

```python
# استفاده از سیستم پرداخت
processor = PaymentProcessor()

# ایجاد پرداخت‌های مختلف
payments = [
    CreditCard(500000, "1234567812345678", "123", "علی محمدی"),
    DigitalWallet(300000, "WALLET_123", 1000000, "زرین‌پال"),
    CryptoPayment(2000000, "Bitcoin", "1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNa")
]

print("💳 سیستم پرداخت چندوجهی")
print("="*60)

# پردازش همه پرداخت‌ها با یک رابط واحد
for payment in payments:
    processor.execute_payment(payment)
    print(f"\n📊 اطلاعات پرداخت: {payment.get_payment_info()}")
    print("-" * 60)

# نمایش تاریخچه
processor.show_payment_history()
```

        1️⃣ وراثت ساده (Single Inheritance)

    <div style="background: linear-gradient(135deg, #667eea10 0%, #f093fb10 100%); border-right: 4px solid #667eea; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        یک کلاس فرزند فقط از <strong>یک</strong> کلاس والد ارث‌بری می‌کند. این ساده‌ترین و رایج‌ترین نوع وراثت است.

```python
class Person:
    """کلاس پایه شخص"""
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        return f"من {self.name} هستم و {self.age} سال دارم"


class Student(Person):
    """کلاس دانشجو که از Person ارث‌بری می‌کند"""
    
    def __init__(self, name, age, student_id, major):
        super().__init__(name, age)
        self.student_id = student_id
        self.major = major
    
    def study(self):
        print(f"{self.name} در حال مطالعه {self.major} است")
    
    def introduce(self):
        """بازنویسی متد معرفی"""
        base_intro = super().introduce()
        return f"{base_intro} و دانشجوی {self.major} با کد {self.student_id} هستم"


# استفاده از وراثت ساده
student = Student("سارا", 20, "98123456", "مهندسی کامپیوتر")
print(student.introduce())
student.study()
```

        2️⃣ وراثت چندسطحی (Multilevel Inheritance)
    </h3>
    <div style="background: linear-gradient(135deg, #667eea10 0%, #f093fb10 100%); border-right: 4px solid #667eea; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        یک کلاس از کلاس دیگری ارث‌بری می‌کند که خود آن از کلاس سومی ارث‌بری کرده است. این ایجاد یک سلسله مراتب می‌کند.
    </div>

```python
class Vehicle:
    """کلاس پایه وسیله نقلیه"""
    
    def __init__(self, brand, fuel_type):
        self.brand = brand
        self.fuel_type = fuel_type
    
    def start_engine(self):
        print(f"موتور {self.brand} روشن شد")
    
    def get_info(self):
        return f"برند: {self.brand}, نوع سوخت: {self.fuel_type}"


class Car(Vehicle):
    """کلاس ماشین که از Vehicle ارث‌بری می‌کند"""
    
    def __init__(self, brand, fuel_type, model, doors):
        super().__init__(brand, fuel_type)
        self.model = model
        self.doors = doors
    
    def honk(self):
        print("بوق بوق!")
    
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, مدل: {self.model}, درها: {self.doors}"


class ElectricCar(Car):
    """کلاس ماشین الکتریکی که از Car ارث‌بری می‌کند"""
    
    def __init__(self, brand, model, doors, battery_capacity, range_km):
        super().__init__(brand, "برقی", model, doors)
        self.battery_capacity = battery_capacity
        self.range_km = range_km
    
    def charge(self):
        print(f"در حال شارژ باتری {self.battery_capacity}kWh")
    
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, ظرفیت باتری: {self.battery_capacity}kWh, برد: {self.range_km}km"


# استفاده از وراثت چندسطحی
tesla = ElectricCar("Tesla", "Model 3", 4, 75, 450)
print(tesla.get_info())
tesla.start_engine()
tesla.honk()
tesla.charge()
```

        3️⃣ وراثت چندگانه (Multiple Inheritance)

    <div style="background: linear-gradient(135deg, #667eea10 0%, #f093fb10 100%); border-right: 4px solid #667eea; border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        یک کلاس از چندین کلاس والد ارث‌بری می‌کند. این امکان قدرتمند است اما باید با احتیاط استفاده شود.
    </div>

```python
class Flyer:
    """کلاس برای موجودات پرنده"""
    
    def fly(self):
        print("در حال پرواز")
    
    def land(self):
        print("فرود آمد")


class Swimmer:
    """کلاس برای موجودات شناگر"""
    
    def swim(self):
        print("در حال شنا")
    
    def dive(self):
        print("شیرجه زد")


class Walker:
    """کلاس برای موجودات راه رونده"""
    
    def walk(self):
        print("در حال راه رفتن")
    
    def run(self):
        print("در حال دویدن")


class Duck(Flyer, Swimmer, Walker):
    """کلاس اردک که از چندین کلاس ارث‌بری می‌کند"""
    
    def __init__(self, name):
        self.name = name
    
    def quack(self):
        print(f"{self.name}: قاق قاق!")
    
    def daily_routine(self):
        """برنامه روزانه اردک"""
        print(f"\n🐥 برنامه روزانه {self.name}:")
        self.walk()
        self.quack()
        self.swim()
        self.fly()
        self.land()
        self.dive()
        self.run()


# استفاده از وراثت چندگانه
donald = Duck("دونالد")
donald.daily_routine()

# بررسی ترتیب ارث‌بری (MRO - Method Resolution Order)
print(f"\n🔍 ترتیب ارث‌بری: {Duck.__mro__}")
```

            <li>سلسله مراتب پیچیده می‌شود</li>
            <li>درک روابط بین کلاس‌ها سخت می‌شود</li>
        </ul>
        <strong>راه حل:</strong> در بسیاری موارد، استفاده از Composition (ترکیب) بهتر از وراثت چندگانه است.

        🎓 کلاس‌های انتزاعی
    </div>
    <div style="background: linear-gradient(135deg, #667eea10 0%, #f093fb10 100%); border-right: 4px solid #667eea; 
                border-radius: 15px; padding: 20px; margin: 15px 0; line-height: 1.8;">
        کلاس‌های انتزاعی کلاس‌هایی هستند که نمی‌توان از آن‌ها شیء مستقیم ساخت و فقط برای ارث‌بری استفاده می‌شوند. 
        این کلاس‌ها معمولاً شامل متدهای انتزاعی هستند که باید در کلاس‌های فرزند پیاده‌سازی شوند.
    </div>
</div>

```python
from abc import ABC, abstractmethod

class DatabaseConnection(ABC):
    """کلاس انتزاعی برای اتصال به پایگاه داده"""
    
    def __init__(self, host, port):
        self.host = host
        self.port = port
        self.is_connected = False
    
    @abstractmethod
    def connect(self):
        """متد انتزاعی برای اتصال به دیتابیس"""
        pass
    
    @abstractmethod
    def disconnect(self):
        """متد انتزاعی برای قطع اتصال"""
        pass
    
    @abstractmethod
    def execute_query(self, query):
        """متد انتزاعی برای اجرای کوئری"""
        pass
    
    def get_connection_info(self):
        """متد عادی - نیازی به پیاده‌سازی در فرزند ندارد"""
        status = "متصل" if self.is_connected else "قطع"
        return f"{self.host}:{self.port} - وضعیت: {status}"


class MySQLConnection(DatabaseConnection):
    """پیاده‌سازی برای MySQL"""
    
    def connect(self):
        print(f"اتصال به MySQL در {self.host}:{self.port}")
        self.is_connected = True
    
    def disconnect(self):
        print("قطع اتصال از MySQL")
        self.is_connected = False
    
    def execute_query(self, query):
        if self.is_connected:
            print(f"اجرای کوئری MySQL: {query}")
            return "نتایج کوئری"
        else:
            raise Exception("ابتدا به دیتابیس متصل شوید")


class PostgreSQLConnection(DatabaseConnection):
    """پیاده‌سازی برای PostgreSQL"""
    
    def connect(self):
        print(f"اتصال به PostgreSQL در {self.host}:{self.port}")
        self.is_connected = True
    
    def disconnect(self):
        print("قطع اتصال از PostgreSQL")
        self.is_connected = False
    
    def execute_query(self, query):
        if self.is_connected:
            print(f"اجرای کوئری PostgreSQL: {query}")
            return "نتایج کوئری"
        else:
            raise Exception("ابتدا به دیتابیس متصل شوید")


# استفاده از کلاس‌های انتزاعی
print("🗄️ سیستم مدیریت پایگاه داده")
print("=" * 50)

mysql = MySQLConnection("localhost", 3306)
postgres = PostgreSQLConnection("localhost", 5432)

# این خط خطا می‌دهد چون نمی‌توان از کلاس انتزاعی شیء ساخت:
# db = DatabaseConnection("localhost", 1234)

connections = [mysql, postgres]

for connection in connections:
    print(f"\n{connection.__class__.__name__}:")
    print(connection.get_connection_info())
    connection.connect()
    connection.execute_query("SELECT * FROM users;")
    connection.disconnect()
```

        حالا بیایید یک سیستم فروشگاه کامل پیاده‌سازی کنیم که از همه مفاهیم یادگرفته شده استفاده می‌کند.

```python
from abc import ABC, abstractmethod

class Product(ABC):
    """کلاس پایه انتزاعی برای همه محصولات"""
    
    def __init__(self, name, base_price):
        self.name = name
        self.base_price = base_price
    
    @abstractmethod
    def calculate_price(self):
        """محاسبه قیمت نهایی محصول"""
        pass
    
    @abstractmethod
    def get_description(self):
        """دریافت توضیحات محصول"""
        pass
    
    def display_info(self):
        """نمایش اطلاعات محصول"""
        print(f"📦 {self.get_description()}")


class PhysicalProduct(Product):
    """محصول فیزیکی - باید حمل شود"""
    
    def __init__(self, name, base_price, weight, dimensions):
        super().__init__(name, base_price)
        self.weight = weight  # وزن به کیلوگرم
        self.dimensions = dimensions  # ابعاد
    
    def calculate_shipping(self):
        """محاسبه هزینه حمل"""
        return self.weight * 10000  # 10,000 تومان به ازای هر کیلوگرم
    
    def calculate_price(self):
        """محاسبه قیمت نهایی با احتساب حمل"""
        return self.base_price + self.calculate_shipping()
    
    def get_description(self):
        """توضیحات محصول فیزیکی"""
        shipping_cost = self.calculate_shipping()
        return (f"{self.name} - قیمت پایه: {self.base_price:,} تومان | "
                f"هزینه حمل: {shipping_cost:,} تومان | "
                f"قیمت نهایی: {self.calculate_price():,} تومان")


class DigitalProduct(Product):
    """محصول دیجیتال - بدون هزینه ارسال"""
    
    def __init__(self, name, base_price, file_size, download_link):
        super().__init__(name, base_price)
        self.file_size = file_size  # حجم فایل به مگابایت
        self.download_link = download_link
    
    def calculate_price(self):
        """محاسبه قیمت نهایی - بدون هزینه اضافه"""
        return self.base_price
    
    def get_description(self):
        """توضیحات محصول دیجیتال"""
        return (f"{self.name} - قیمت: {self.base_price:,} تومان | "
                f"حجم: {self.file_size}MB | لینک دانلود: {self.download_link}")
    
    def download(self):
        """متد اختصاصی برای دانلود"""
        print(f"📥 در حال دانلود {self.name}...")
        print(f"🔗 لینک: {self.download_link}")


class SubscriptionProduct(Product):
    """محصول اشتراکی"""
    
    def __init__(self, name, base_price, duration_months, auto_renew):
        super().__init__(name, base_price)
        self.duration_months = duration_months
        self.auto_renew = auto_renew
    
    def calculate_monthly_price(self):
        """محاسبه قیمت ماهانه"""
        return self.base_price / self.duration_months
    
    def calculate_price(self):
        """محاسبه قیمت نهایی"""
        return self.base_price
    
    def get_description(self):
        """توضیحات محصول اشتراکی"""
        monthly_price = self.calculate_monthly_price()
        renew_status = "فعال" if self.auto_renew else "غیرفعال"
        return (f"{self.name} - قیمت کل: {self.base_price:,} تومان | "
                f"مدت: {self.duration_months} ماه | "
                f"قیمت ماهانه: {monthly_price:,.0f} تومان | "
                f"تمدید خودکار: {renew_status}")
    
    def activate_subscription(self):
        """فعال‌سازی اشتراک"""
        print(f"✅ اشتراک {self.name} برای {self.duration_months} ماه فعال شد")


class ShoppingCart:
    """سبد خرید - نمایش چندریختی"""
    
    def __init__(self):
        self.items = []
    
    def add_item(self, product):
        """اضافه کردن محصول به سبد"""
        self.items.append(product)
        print(f"✅ {product.name} به سبد خرید اضافه شد")
    
    def remove_item(self, product_name):
        """حذف محصول از سبد"""
        for item in self.items:
            if item.name == product_name:
                self.items.remove(item)
                print(f"❌ {product_name} از سبد خرید حذف شد")
                return
        print(f"⚠️ محصول {product_name} در سبد یافت نشد")
    
    def calculate_total(self):
        """محاسبه جمع کل سبد"""
        return sum(item.calculate_price() for item in self.items)
    
    def show_cart(self):
        """نمایش محتویات سبد"""
        print("\n" + "="*70)
        print("🛒 سبد خرید")
        print("="*70)
        
        if not self.items:
            print("سبد خرید خالی است")
            return
        
        for i, item in enumerate(self.items, 1):
            print(f"{i}. {item.get_description()}")
        
        print("-"*70)
        print(f"💰 جمع کل: {self.calculate_total():,} تومان")
        print("="*70)
    
    def checkout(self):
        """تسویه حساب"""
        if not self.items:
            print("❌ سبد خرید خالی است")
            return
        
        print("\n" + "🎊 تسویه حساب")
        print("="*70)
        
        total = self.calculate_total()
        print(f"مبلغ قابل پرداخت: {total:,} تومان")
        
        # انجام عملیات خاص برای هر نوع محصول
        for item in self.items:
            if isinstance(item, DigitalProduct):
                item.download()
            elif isinstance(item, SubscriptionProduct):
                item.activate_subscription()
        
        print("✅ پرداخت با موفقیت انجام شد")
        print("🙏 از خرید شما متشکریم")
        print("="*70)
        
        # خالی کردن سبد بعد از پرداخت
        self.items = []
```

```python
# استفاده از سیستم فروشگاه
cart = ShoppingCart()

# ایجاد محصولات مختلف
products = [
    PhysicalProduct("لپتاپ دل", 25000000, 2.5, "40x30x5 cm"),
    DigitalProduct("کتاب الکترونیکی پایتون", 50000, 15, "https://example.com/python-book"),
    SubscriptionProduct("اشتراک طلایی Netflix", 300000, 6, True),
    PhysicalProduct("ماوس گیمینگ", 800000, 0.2, "12x6x3 cm"),
    DigitalProduct("دوره آموزشی هوش مصنوعی", 1200000, 450, "https://example.com/ai-course")
]

print("🏪 فروشگاه آنلاین")
print("="*70)

# نمایش همه محصولات
print("\n📋 لیست محصولات:")
print("-"*70)
for i, product in enumerate(products, 1):
    product.display_info()

# اضافه کردن محصولات به سبد
print("\n" + "🛒 افزودن محصولات به سبد:")
print("-"*70)
for product in products[:3]:  # فقط 3 محصول اول
    cart.add_item(product)

# نمایش سبد خرید
cart.show_cart()

# حذف یک محصول
print("\n" + "🗑️ حذف محصول از سبد:")
print("-"*70)
cart.remove_item("ماوس گیمینگ")

# نمایش سبد خرید بعد از حذف
cart.show_cart()

# تسویه حساب
cart.checkout()

# نمایش سبد خرید بعد از تسویه
cart.show_cart()
```
