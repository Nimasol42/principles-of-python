  <h3>Python Interpreter و محیط توسعه</h3>
  <p>
    برای نوشتن و <strong>اجرای</strong> برنامه‌های پایتون، کافی است <strong>Python Interpreter</strong> را نصب کنید. برخلاف برخی زبان‌ها، پایتون نیازی به مرحله جداگانه <strong>کامپایل</strong> ندارد و کدها مستقیماً توسط مفسر اجرا می‌شوند.
  </p>
  <ul>
    <li><strong>Interpreter</strong> — اجرای مستقیم کدهای <code>.py</code> و تبدیل آن‌ها به دستوراتی که سیستم قادر به فهم آن‌هاست.</li>
    <li><strong>Interactive Shell</strong> — محیط تعاملی برای آزمایش سریع دستورات و کدهای کوتاه.</li>
    <li><strong>Standard Library</strong> — مجموعه گسترده‌ای از ماژول‌ها و توابع آماده برای کار با فایل‌ها، شبکه، داده‌ها و بسیاری کاربردهای دیگر.</li>
    <li><strong>Documentation</strong> — مستندات رسمی پایتون و راهنمای کتابخانه‌ها برای یادگیری و توسعه سریع‌تر.</li>
    <li><strong>IDE یا ویرایشگر کد</strong> — ابزارهایی مانند PyCharm، VS Code یا حتی IDLE که همراه پایتون نصب می‌شود، برای نوشتن و مدیریت پروژه‌ها.</li>
  </ul>
  <div class="slide-note">
    جمع‌بندی: برای شروع کار با پایتون، نصب <strong>Interpreter</strong> و آشنایی با <strong>کتابخانه استاندارد</strong> کافی است؛ سپس می‌توانید با یک ویرایشگر مناسب، کدنویسی را آغاز کنید.
  </div>

  <h3>خصوصیات زبان پایتون</h3>
  <ul>
    <li>
      <strong>سادگی:</strong>

    </li>
    <li>
      <strong>شیءگرایی (Object-Oriented):</strong>
      پشتیبانی کامل از کلاس‌ها و اشیاء برای مدل‌سازی مفاهیم دنیای واقعی، افزایش تعامل بین اجزا و <strong>استفاده‌مجدد از کد</strong>.
    </li>
    <div class="images">
      <img src="img/oop.jpg" alt="Python OOP">
    </div>
    <li>
      <strong>قابل‌حمل و چندسکویی:</strong>
      کد پایتون روی سیستم‌عامل‌های مختلف (Windows، Linux، macOS) بدون تغییر اجرا می‌شود.
    </li>
    <li>
      <strong>تفسیر مستقیم:</strong>
      کد منبع پایتون مستقیماً توسط <strong>مفسر پایتون</strong> اجرا می‌شود و نیازی به مرحله جداگانه کامپایل ندارد. این ویژگی توسعه و آزمایش سریع را ممکن می‌سازد.
    </li>
    <div class="images">
      <img src="img/1.png" alt="Python Cross-Platform">
    </div>
    <li>
      <strong>چندریسمانی و چندپردازشی:</strong>
      امکان اجرای همزمان چند وظیفه با استفاده از Threadها یا Processها، مناسب برای پردازش داده، کارهای شبکه‌ای و برنامه‌های واکنش‌گرا.
    </li>
    <li>
      <strong>امن و پایدار (Secure & Robust):</strong>
      مدیریت خطا با سیستم استثناها (Exceptions)، کتابخانه‌های امن برای کار با داده‌ها و جلوگیری از آسیب‌پذیری‌های رایج.
    </li>
  </ul>
