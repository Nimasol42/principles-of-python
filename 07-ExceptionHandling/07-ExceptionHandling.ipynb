  <h2>مقدمه</h2>
  <p>
    در مسیر توسعه‌ی نرم‌افزار، خطاها بخش جدایی‌ناپذیر برنامه‌نویسی هستند. گاهی ورودی کاربر نامعتبر است، گاهی فایل موردنظر وجود ندارد، و گاهی عملیات ریاضی مثل تقسیم بر صفر باعث توقف برنامه می‌شود. اگر این خطاها به‌درستی مدیریت نشوند، می‌توانند منجر به رفتارهای غیرمنتظره، از دست رفتن داده‌ها، یا تجربه‌ی کاربری ضعیف شوند.
  </p>
  <p>
    <strong>استثناء (Exception)</strong> به شرایطی گفته می‌شود که در زمان اجرای برنامه رخ می‌دهد و اجرای عادی را مختل می‌کند. برخلاف <strong>خطاهای نحوی (Syntax Errors)</strong> که مانع اجرای برنامه می‌شوند، استثناءها قابل کنترل هستند و می‌توان با استفاده از ساختارهای مناسب، از توقف ناگهانی برنامه جلوگیری کرد.
  </p>
  <p>
    مدیریت استثناء به ما این امکان را می‌دهد که:
    <ul>
      <li>رفتار برنامه را در شرایط غیرعادی کنترل کنیم</li>
      <li>منابع را به‌درستی آزاد کنیم (مثل بستن فایل‌ها یا قطع اتصال پایگاه داده)</li>
      <li>پیام‌های مناسب و قابل‌فهم به کاربر نمایش دهیم</li>
    </ul>
  </p>

</div>

<div class="card bold" dir="rtl">
  <h2>تعریف  استثناء و خطا</h2>
  <p>
    در علوم رایانه، <strong>خطا (Error)</strong> به هرگونه نقص در منطق یا ساختار برنامه گفته می‌شود که مانع اجرای صحیح آن می‌شود. اما در پایتون، خطاها به دو دسته‌ی اصلی تقسیم می‌شوند:
    <ul>
      <li><strong>Syntax Errors</strong>: خطاهای نحوی که در مرحله‌ی ترجمه (Parsing) رخ می‌دهند و مانع اجرای برنامه می‌شوند.</li>
      <li><strong>Exceptions</strong>: شرایط غیرعادی در زمان اجرا که می‌توان آن‌ها را مدیریت کرد.</li>
    </ul>
    <p>
    استثناءها ابزارهایی هستند که پایتون برای کنترل رفتار غیرمنتظره در زمان اجرا فراهم کرده است.</p>
  </p>
</div>

<div class="card bold" dir="rtl">
  <h2>استثناءها چگونه کار می‌کنند؟</h2>
  <p>
    وقتی در زمان اجرای برنامه شرایطی غیرعادی رخ دهد، پایتون یک شیء از کلاس استثناء مربوطه ایجاد می‌کند و آن را «پرتاب» (raise) می‌کند. اگر این استثناء توسط برنامه‌نویس «گرفته» (catch) نشود، برنامه متوقف می‌شود و پیام خطا نمایش داده می‌شود.
  </p>
  <p>
    برای جلوگیری از توقف برنامه، باید از ساختار <code>try / except</code> استفاده کنیم تا استثناء را گرفته و واکنش مناسبی نشان دهیم.
  </p>
</div>

    <ul>
      <li><code>ValueError</code>: زمانی که مقدار نامعتبر به تابع داده شود.</li>

  <h2>تفاوت BaseException و Exception</h2>
  <p>
    تفاوت اصلی بین <code>BaseException</code> و <code>Exception</code> در دامنه‌ی کاربرد آن‌هاست:
    <ul>
      <li><strong>BaseException</strong>: شامل تمام خطاها، از جمله خطاهای سیستمی که معمولاً نباید گرفته شوند.</li>
      <li><strong>Exception</strong>: شامل خطاهای منطقی و قابل‌مدیریت که در برنامه‌های معمول رخ می‌دهند.</li>
    </ul>
  </p>

  </p>
  <p>
    در مقابل، گرفتن <code>Exception</code> فقط خطاهای منطقی را مدیریت می‌کند و رفتار برنامه را قابل پیش‌بینی‌تر و ایمن‌تر می‌سازد.

# Recommended: catch logical errors only
try:
    risky_operation()
except Exception as e:
    print("Handled:", e)
# Not recommended: catching system-level exceptions
try:
    risky_operation()
except BaseException as e:
    print("Dangerous catch:", e)

```python
# Custom exception in input validation
class InvalidAgeError(Exception):
    def __init__(self, age):
        self.age = age
        super().__init__(f"Invalid age: {age}. Age must be between 0 and 120.")
def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError(age)
    return True
try:
    user_age = int(input("Enter your age: "))
    validate_age(user_age)
except InvalidAgeError as e:
    print("Validation failed:", e)
else:
    print("Age is valid.")
```
