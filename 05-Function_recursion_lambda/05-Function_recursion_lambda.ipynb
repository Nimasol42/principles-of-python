<div class="card" dir="rtl">
  <h2>اهمیت تابع</h2>
  <p>
    در برنامه‌نویسی، توابع مانند جعبه‌ابزارهایی هستند که کارهای مشخص را برای ما انجام می‌دهند.
    به جای تکرار چندین‌باره یک قطعه کد، می‌توانیم آن را یک‌بار در قالب یک تابع بنویسیم و هر زمان که لازم شد، فراخوانی کنیم.
  </p>
  <p>
    این کار باعث می‌شود کد ما <strong>خواناتر</strong>، <strong>کوتاه‌تر</strong> و <strong>قابل نگهداری‌تر</strong> باشد.
  </p>
  <p>
    تصور کنید در یک فروشگاه آنلاین، محاسبه هزینه نهایی سفارش در بخش‌های مختلف سایت لازم است.
    اگر این محاسبه را در قالب یک تابع بنویسیم، هر بار که نیاز داشتیم، فقط آن تابع را صدا می‌زنیم.
  </p>
  <p>
    به این ترتیب، اگر روزی بخواهیم فرمول محاسبه را تغییر دهیم، کافی است فقط همان تابع را ویرایش کنیم
    و همه بخش‌های برنامه به‌طور خودکار به‌روزرسانی می‌شوند.
  </p>
</div>

  <h2>پارامترهای موقعیتی (Positional Parameters)</h2>
  <pre style="background-color: #272822; font-family: Consolas, monospace; color: #f8f8f2; padding: 12px; border-radius: 8px; direction: ltr; text-align: left; line-height: 1.6;">

```python
def say_hello():
    print("Hello World")

say_hello()
```

def welcome():
    print("Welcome to Python")

welcome()

```python
def multiply(a, b):
    return a * b

print(multiply(4, 5))
```

```python
def greet(name):
    return f"Hello {name}"

print(greet("Ali"))
```

```python
def greet_user(name="Guest"):
    return f"Hello {name}"

print(greet_user())
```

```python
def power(base, exponent=2):
    return base ** exponent

print(power(5))
```

```python
def min_max(numbers):
    return min(numbers), max(numbers)

print(min_max([3, 7, 1, 9]))
```

```python
def divide_and_remainder(a, b):
    return a // b, a % b

print(divide_and_remainder(10, 3))
```

```python
def sum_all(*args):
    return sum(args)

print(sum_all(1, 2, 3, 4))
```

```python
def multiply_all(*args):
    result = 1
    for num in args:
        result *= num
    return result

print(multiply_all(2, 3, 4))
```

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Ali", age=30)
```

```python
def build_dict(**kwargs):
    return kwargs

print(build_dict(a=1, b=2, c=3))
```

```python
def square(x):
    return x * x

def sum_of_squares(a, b):
    return square(a) + square(b)

print(sum_of_squares(3, 4))
```

```python
def double(x):
    return x * 2

def quadruple(y):
    return double(double(y))

print(quadruple(5))
```

    با استفاده از کلمه کلیدی <code>global</code> می‌توانیم از داخل تابع به متغیر سراسری دسترسی پیدا کرده و مقدار آن را تغییر دهیم. بدون این دستور، ایجاد متغیر همنام در تابع یک متغیر محلی جدید می‌سازد و نسخه سراسری بدون تغییر می‌ماند.

<!-- استفاده از nonlocal -->
<div class="card" dir="rtl">
  <h2>استفاده از nonlocal</h2>
  <pre style="background-color: #272822; font-family: Consolas, monospace; color: #f8f8f2; padding: 12px; border-radius: 8px; direction: ltr; text-align: left; line-height: 1.6;">
def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
        return count
    print(inner())
    print(inner())
outer()

  </p>

  <h2>تابع باگشتی  </h2>
  <p style="background-color: #f9f9f9; padding: 8px; border-radius: 6px;">
    تابع بازگشتی (Recursive Function) تابعی است که در تعریف خود، خودش را فراخوانی می‌کند. ایده اصلی این است که یک مسئله بزرگ را به مسائل کوچک‌تر و مشابه تقسیم کنیم تا حل آن ساده‌تر شود. این فرآیند تا زمانی ادامه پیدا می‌کند که به یک حالت پایه (شرط توقف) برسیم که دیگر نیازی به فراخوانی مجدد تابع نباشد.
    <br><br>
    استفاده از بازگشت زمانی مناسب است که:
    <ul>
      <li>مسئله ماهیت تکرارشونده یا خود-مشابه دارد (مثل ساختار درختی یا دنباله‌ها).</li>
      <li>راه‌حل طبیعی آن به صورت تقسیم مسئله به زیرمسئله‌ها بیان می‌شود.</li>
      <li>کد بازگشتی خواناتر و ساده‌تر از استفاده از حلقه‌ها باشد.</li>
    </ul>
    <p>
    منطق کلی بازگشت:
    </p>
      <li>بررسی شرط توقف (Base Case) — نقطه‌ای که بازگشت متوقف می‌شود.</li>
      <li>تقسیم مسئله به نسخه کوچک‌تر.</li>
      <li>فراخوانی مجدد تابع با ورودی کوچک‌تر.</li>
    </ol>
  </p>

    <ul>
      <li>محاسبه دنباله‌ها (مثل فیبوناچی).</li>
      <li>پیمایش ساختارهای درختی و گراف‌ها.</li>
      <li>حل مسائل ریاضی مانند فاکتوریل، باینری سرچ، و Tower of Hanoi.</li>

def recursive_function(parameters):

    هر تابع بازگشتی باید دو بخش اصلی داشته باشد:
    <strong>شرط توقف</strong> (Base Case) که جلوی بی‌نهایت شدن بازگشت را می‌گیرد، و <strong>فراخوانی بازگشتی</strong> که مسئله را به نسخه کوچک‌تر تبدیل می‌کند.

def factorial(n):

      <li>در مسائلی که بازگشت پیچیدگی غیرضروری ایجاد می‌کند، از حلقه‌ها استفاده کنید.</li>

    برنامه‌ای بنویسید که با استفاده از تابع بازگشتی، فاکتوریل یک عدد صحیح غیرمنفی را محاسبه کند.  
    فاکتوریل n برابر است با حاصل‌ضرب تمام اعداد صحیح مثبت کوچکتر یا مساوی n.

<div class="card" dir="rtl">
  <h2>مثال ۵: جستجوی بازگشتی در لیست</h2>
  <p style="background-color: #f9f9f9; padding: 8px; border-radius: 6px;">
    برنامه‌ای بنویسید که با استفاده از تابع بازگشتی، بررسی کند آیا یک عدد در لیست وجود دارد یا خیر.  
    <strong>شرط توقف:</strong> اگر لیست خالی باشد، نتیجه False است.
  </p>

```python
def search_list(numbers, target):
    if not numbers:
        return False
    if numbers[0] == target:
        return True
    return search_list(numbers[1:], target)

print(search_list([2, 4, 6, 8], 6))
print(search_list([2, 4, 6, 8], 5))
```
