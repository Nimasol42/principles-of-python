  <p>

  </p>
  <p>
    عملیات <strong>I/O</strong> یا <strong>Input/Output</strong> به فرآیند دریافت داده از منابع خارجی (مانند فایل‌ها، کیبورد، یا شبکه) و ارسال داده به آن‌ها اشاره دارد. در این نوت‌بوک، با مفاهیم پایه‌ای کار با فایل‌ها در پایتون آشنا می‌شویم و یاد می‌گیریم چگونه فایل‌ها را بخوانیم، بنویسیم، و به‌صورت امن مدیریت کنیم.
  </p>

<div class="card bold" dir="rtl">
  <h2>تفاوت بین حافظه RAM و حافظه دائمی</h2>
  <p>
    <strong>RAM (حافظه فرّار):</strong> جایی است که برنامه هنگام اجرا داده‌ها و دستورها را نگه می‌دارد. بسیار سریع است (تاخیر در حد نانوثانیه)،
    اما با پایان اجرای برنامه یا قطع برق، محتوا پاک می‌شود. RAM برای محاسبات لحظه‌ای، بافرها و ساختارهای داده درون‌حافظه‌ای ایده‌آل است.
  </p>
  <p>
    <strong>حافظه دائمی (Disk/SSD):</strong> رسانه‌ای پایدار با ظرفیت بالا و تاخیر بیشتر (میلی‌ثانیه) که داده‌ها را در قالب فایل‌ها ذخیره می‌کند.
    هدف آن نگه‌داری بلندمدت، اشتراک‌گذاری بین اجراها و انتقال بین سیستم‌هاست. فایل‌ها روی سیستم‌عامل در قالب مسیرها، مجوزها و فرمت‌های مختلف مدیریت می‌شوند.
  </p>
  <ul>
    <li><strong>تاخیر و پهنای‌باند:</strong> RAM بسیار سریع‌تر است؛ Disk کندتر اما پایدار و حجیم.</li>
    <li><strong>دوام:</strong> RAM فرّار؛ Disk پایدار. نتیجه‌ها در Disk قابل بازیابی‌اند.</li>
    <li><strong>الگوی دسترسی:</strong> RAM تصادفی و مستقیم؛ Disk مبتنی بر بلوک‌ها و فایل‌ها، تحت مدیریت سیستم‌فایل.</li>
  </ul>

  <h2>انواع فایل‌ها: متنی (Text) و باینری (Binary)</h2>
  <p>
    فایل‌ها را می‌توان به دو دسته اصلی تقسیم کرد: <strong>متنی</strong> و <strong>باینری</strong>.
    این تقسیم‌بندی بر اساس نحوه ذخیره‌سازی و خوانایی داده‌ها برای انسان انجام می‌شود.
  </p>
  <p>
    <strong>فایل‌های متنی (Text):</strong> شامل کاراکترهای قابل خواندن برای انسان هستند. برای مثال:
    <code>.txt</code>، <code>.csv</code>، <code>.json</code>. این فایل‌ها را می‌توان با هر ویرایشگر متنی ساده باز کرد.
    نکته مهم در فایل‌های متنی، <em>Encoding</em> است (مانند UTF-8 یا ASCII) که تعیین می‌کند کاراکترها چگونه به بایت‌ها تبدیل شوند.
  </p>
  <p>
    <strong>فایل‌های باینری (Binary):</strong> داده‌ها را به صورت خام و غیرقابل‌خواندن برای انسان ذخیره می‌کنند.
    نمونه‌ها: تصاویر (<code>.jpg</code>، <code>.png</code>)، فایل‌های صوتی (<code>.mp3</code>)، ویدیوها (<code>.mp4</code>) و آرشیوها (<code>.zip</code>).
    این فایل‌ها معمولاً دارای ساختار داخلی پیچیده و هدرهای مخصوص هستند.
  </p>

  <p>

  </p>
  <p>
    شیء فایل متدهای متنوعی دارد مانند <code>read</code> برای خواندن، <code>write</code> برای نوشتن و <code>close</code> برای بستن فایل. 
  </p>

<div class="card bold" dir="rtl">
  <h2>خواندن از فایل در پایتون</h2>
  <p>
    یکی از پرکاربردترین عملیات در برنامه‌نویسی، خواندن داده‌ها از فایل است. پایتون سه متد اصلی برای این کار فراهم کرده است:
    <code>read</code>، <code>readline</code> و <code>readlines</code>. هرکدام از این متدها رفتار خاصی دارند و انتخاب درست آن‌ها به نیاز شما و اندازه فایل بستگی دارد.
  </p>
  <p>
    - <strong>read</strong>: کل محتوای فایل یا تعداد مشخصی از کاراکترها را به صورت یک رشته برمی‌گرداند.<br>
    - <strong>readline</strong>: یک خط از فایل را برمی‌گرداند. هر بار فراخوانی، خط بعدی را می‌خواند.<br>
    - <strong>readlines</strong>: تمام خطوط فایل را در قالب یک لیست از رشته‌ها برمی‌گرداند.
  </p>

  <h3>متد read</h3>
  <p>
    این متد کل محتوای فایل را به صورت یک رشته بازمی‌گرداند. اگر آرگومان عددی بدهیم، به همان تعداد کاراکتر خوانده می‌شود. برای فایل‌های کوچک مناسب است.
  </p>
  <pre style="background-color:#272822;font-family:Consolas,monospace;color:#f8f8f2;padding:12px;border-radius:8px;direction:ltr;text-align:left;line-height:1.6;">
f = open("sample.txt", "r")
content = f.read()
print("Full content of file:")
print(content)
f.close()
  </pre>
  <p>
    در این مثال کل محتوای فایل یکجا خوانده و چاپ می‌شود.
  </p>

  <h3>متد readline</h3>
  <p>
    این متد یک خط از فایل را برمی‌گرداند. اگر دوباره فراخوانی شود، خط بعدی خوانده می‌شود. برای پردازش خط به خط بسیار مناسب است.
  </p>
  <pre style="background-color:#272822;font-family:Consolas,monospace;color:#f8f8f2;padding:12px;border-radius:8px;direction:ltr;text-align:left;line-height:1.6;">
f = open("sample.txt", "r")
line1 = f.readline()
line2 = f.readline()
print("First line:", line1.strip())
print("Second line:", line2.strip())
f.close()
  </pre>
  <p>
    در این مثال دو خط اول فایل خوانده و چاپ می‌شود. متد <code>strip</code> برای حذف کاراکترهای اضافی انتهای خط استفاده شده است.
  </p>
  <pre style="background-color:#272822;font-family:Consolas,monospace;color:#f8f8f2;padding:12px;border-radius:8px;direction:ltr;text-align:left;line-height:1.6;">

  <p>
    در پایتون می‌توان از ساختار <code>with</code> برای مدیریت فایل‌ها استفاده کرد. این روش یک <strong>context manager</strong> است که وظیفه باز و بسته کردن فایل را به صورت خودکار بر عهده دارد.
  </p>
  <p>
    مزایای استفاده از context manager:
    <ul>
      <li>جلوگیری از فراموشی فراخوانی <code>close</code> پس از پایان کار</li>
      <li>مدیریت خودکار منابع سیستم</li>
      <li>کد خواناتر و کوتاه‌تر</li>
    </ul>
  </p>
  <pre style="background-color:#272822;font-family:Consolas,monospace;color:#f8f8f2;padding:12px;border-radius:8px;direction:ltr;text-align:left;line-height:1.6;">
with open("sample.txt", "r") as f:
    for line in f:
        print(line.strip())

  <p>
    در این روش نیازی به فراخوانی <code>close</code> نیست، زیرا فایل به صورت خودکار پس از خروج از بلاک بسته می‌شود.

      <li>از بروز خطاهای احتمالی هنگام دسترسی مجدد به فایل جلوگیری می‌شود.</li>
    </ul>
  </p>

  <p>
    JSON به‌صورت پیش‌فرض اشیای سفارشی، تاریخ/زمان یا مجموعه‌ها را پشتیبانی نمی‌کند. می‌توان با پارامتر <code>default</code> در <code>dumps</code>/<code>dump</code> تبدیل سفارشی تعریف کرد. در این تبدیل معمولاً شیٔ به رشته یا دیکشنری قابل‌سریال‌سازی تبدیل می‌شود؛ در هنگام بارگذاری باید بازیابی معکوس را خودتان انجام دهید.

import json, datetime
def encode_custom(obj):
    if isinstance(obj, datetime.datetime):
        return {"__type": "datetime", "iso": obj.isoformat()}

payload = {"when": datetime.datetime(2025, 10, 4, 20, 45), "tags": {"a", "b"}}
text = json.dumps(payload, default=encode_custom, ensure_ascii=False, indent=2)
print(text)
