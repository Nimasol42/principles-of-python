  <h2>معرفی حلقه while</h2>
  <p>
    حلقه <strong>while</strong> یکی از ساختارهای تکرار در برنامه‌نویسی است که تا زمانی که یک شرط برقرار باشد، 
    دستورات داخل خود را اجرا می‌کند.  
    به عبارت دیگر، تا وقتی شرط <em>True</em> باشد، حلقه ادامه پیدا می‌کند و به محض اینکه شرط <em>False</em> شود، متوقف می‌شود.
  </p>

  <p>
    ساختار کلی حلقه while در پایتون به این شکل است:
  </p>

  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
count = 0
while count &lt; 100:
    print("Welcome to Python!")
    count += 1
  </pre>

  <p>
    در این مثال:
    <ul>
      <li>متغیر <code>count</code> از مقدار ۰ شروع می‌شود،</li>
      <li>تا زمانی که <code>count &lt; 100</code> باشد، متن چاپ می‌شود،</li>
      <li>در هر بار اجرای حلقه، <code>count</code> یک واحد افزایش می‌یابد،</li>
      <li>وقتی <code>count</code> به ۱۰۰ برسد، شرط برقرار نیست و حلقه متوقف می‌شود.</li>
    </ul>
  </p>

  <hr class="divider" />

  <h3> مثال  </h3>
  <p>
    تصور کنید در سرزمین افسانه‌ای «پایتونیا»، یک جادوگر مهربان به نام «پای‌مستر» زندگی می‌کند.  
    او یک طوطی سخنگو دارد که باید ۱۰ بار جمله‌ی جادویی <strong>"دروازه‌ها باز شوند!"</strong> را بگوید تا دروازه‌ی قلعه باز شود.
  </p>

  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
parrot_count = 0
while parrot_count &lt; 10:
    print("Open the gates!")
    parrot_count += 1

print("🎉 The castle gates are now open! Welcome to Pythonia!")
  </pre>

  <p>
    در این داستان:
    <ul>
      <li>حلقه while نقش «تکرار طلسم» را دارد،</li>
      <li>شرط ادامه، تعداد دفعاتی است که طوطی باید جمله را بگوید،</li>
      <li>وقتی شرط تمام شود، اتفاق جادویی (باز شدن دروازه) رخ می‌دهد.</li>
    </ul>
  </p>



  <hr class="divider" />
</div>

```python
parrot_count = 0
while parrot_count &lt; 10:
    print("Open the gates!")
    parrot_count += 1

print(" The castle gates are now open! Welcome to Pythonia!")
```

  <h3>مثال </h3>
  <p>
    فرض کنید در غار گنج «پایتونیا» هستید و با بیلچه جادویی خود مشغول حفاری هستید.  
    شما تا زمانی که سکه طلایی پیدا نکرده‌اید، به حفاری ادامه می‌دهید.  
    نمی‌دانید چند بار باید بیل بزنید، به همین دلیل حلقه <strong>while</strong> بهترین انتخاب است.
  </p>

  <p>
    در این مثال:
    <ul>
      <li>حلقه تا زمانی که سکه پیدا نشده ادامه دارد،</li>
      <li>شرط توقف، پیدا شدن سکه است،</li>
      <li>این مدل حلقه برای سناریوهایی که پایان کار مشخص نیست، ایده‌آل است.</li>
    </ul>

```python
found_coin = False
dig_count = 0
while not found_coin:
    dig_count += 1
    print(f"Dig number {dig_count}... still searching!")
    if dig_count == 7:
        found_coin = True
print("You found the golden coin!")
```

  <h2 style="margin-bottom: 20px;">مثال : حدس اعداد </h2>
  <p> 
  </p>
  <li style="margin-bottom: 20px;">
    برنامه ای بنویسید که به طور تصادفی یک عدد صحیح بین ۰ تا ۱۰۰ را تولید کند.
  </li>
  <li style="margin-bottom: 20px;">      
    برنامه به کاربر می گوید تا پشت سر هم اعدادی را وارد کند تا زمانی که عدد وارد شده با عدد تصادفی تولید شده برابر شود.
  </li>
  <li style="margin-bottom: 20px;">     
    به ازای هر ورودی کاربر، برنامه به او می گوید آیا مقدار وارد شده کاربر از عدد مربوطه کوچکتر است یا بزرگتر. در نتیجه کاربر میتواند ورودی بعدی را هوشمندانه تر تولید کند.
  </li>

</div>

```python
import random

# Generate a random number to be guessed
number = random.randint(0, 100)

print("Guess a magic number between 0 and 100")

guess = -1
while guess != number:
    # Prompt the user to guess the number
    guess = int(input("\nEnter your guess: "))

    if guess == number:
        print(f"Yes, the number is {number}")
    elif guess > number:
        print(f"Your guess ({guess}) is too high")
    else:
        print(f"Your guess ({guess}) is too low")
```

<div class="card" dir="rtl">
  <h2 style="margin-bottom: 4px;">مثال : ابزار پیشرفته آموزش ریاضی </h2>
  <p> 
  </p>
  <p>
    ابزار یادگیری تفریق در هر اجرا تنها یک سوال مطرح می‌کرد.
  </p>
  <p>     
    با استفاده از <strong>loop</strong> می‌توانیم سوالاتی را پشت سر هم مطرح کنیم.
  </p>
  <li>     
    برنامه‌ای بنویسید که ۵ سوال از کاربر بپرسد و تعداد پاسخ‌های صحیح را پس از پایان سوالات اعلام کند.
  </li>
  <p>
    همچنین لیست سوالات به همراه صحیح یا غلط بودن پاسخ ارائه شود.
  </p>
  <p>
    زمان سپری شده برای پاسخ دانش‌آموز نیز محاسبه شود.
  </p>
  <li>
دستور زیر زمان فعلی را به میلی‌ثانیه برمی‌گرداند:  
  </li>
  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
import time
current_millis = int(time.time() * 1000)
  </pre>
<p>
همچنین این تابع در ماژول <em>time</em> قرار دارد.
</p>

</div>

```python
import random
import time

NUMBER_OF_QUESTIONS = 5  # تعداد سوالات

correct_count = 0  # شمارش پاسخ‌های صحیح
count = 0  # شمارش تعداد سوالات
start_time = int(time.time() * 1000)  # زمان شروع به میلی‌ثانیه
output = ""  # رشتهٔ خروجی

while count < NUMBER_OF_QUESTIONS:
    # 1. تولید دو عدد صحیح تصادفی یک‌رقمی
    number1 = random.randint(0, 9)
    number2 = random.randint(0, 9)

    # 2. اگر number1 < number2، جابه‌جا کردن آن‌ها
    if number1 < number2:
        number1, number2 = number2, number1

    # 3. پرسیدن سوال از کاربر
    answer = int(input(f"What is {number1} - {number2}? "))

    # 4. بررسی پاسخ و نمایش نتیجه
    if number1 - number2 == answer:
        print("You are correct!")
        correct_count += 1
    else:
        print(f"Your answer is wrong.\n{number1} - {number2} should be {number1 - number2}")

    # افزایش شمارنده
    count += 1

    # اضافه کردن نتیجه به خروجی
    output += f"\n{number1}-{number2}={answer} {'correct' if number1 - number2 == answer else 'wrong'}"

end_time = int(time.time() * 1000)  # زمان پایان
test_time = end_time - start_time

print(f"Correct count is {correct_count}")
print(f"Test time is {test_time // 1000} seconds")
print(output)
```

  <h2 style="margin-bottom: 4px;"> مثال:پایان دادن به حلقه با یک مقدار کنترلی </h2>
  <p> 
  </p>
  <p>
    در بیشتر مواقع تعداد دفعات اجرای حلقه از قبل مشخص نمی باشد.
  </p>
  <p>     
    در چنین مواقعی از یک مقدار ورودی برای پایان دادن به حلقه استفاده میکنیم.
  </p>
  <li style="margin-bottom: 50px;">     
  به این مقدار اصطلاحاً <strong> نگهبان (sentinel) </strong> گفته میشود.
  </li>
  <li>
    حال برنامه ای بنویسید که مجموع تعداد نامشخصی عدد را حساب کند.
  </li>
  <p>
    مقدار 0 نگهبان یا پایان دهنده ورودی باشد.
  </p>


</div>

```python
sum_values = 0
data = int(input("Enter an int value (the program exits if the input is 0): "))

while data != 0:
    sum_values += data
    data = int(input("Enter an int value (the program exits if the input is 0): "))

print(f"The sum is {sum_values}")
```

<div class="card" dir="rtl">
  <h2 style="margin-bottom: 4px;"> نکته مهم </h2>
  <p> 
  </p>
  <li>
    از <strong>مقادیر ممیز شناور</strong> (مانند float) برای بررسی <strong>مساوی بودن</strong> در شرط کنترلی حلقه استفاده <strong>نکنید</strong>.
  </li>
  <p>    
    چون برای برخی مقادیر، تقریبی و نه دقیق هستند.
  </p>
  <p>
    در نتیجه ممکن است منجر به خطا و نتایج نادرست شوند.
  </p>
  <li>
    فرض کنید می‌خواهیم حاصل عبارت <strong>1 + 0.9 + ... + 0.2 + 0.1</strong> را محاسبه نماییم‌:
  </li>  
  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
item = 1.0
sum_values = 0.0
while item != 0:  # No guarantee item will be exactly 0
    sum_values += item
    item -= 0.1
print(sum_values)
  </pre>

  <li>
    متغیر <code>item</code> از ۱ شروع به کم شدن ۰.۱ در هر تکرار حلقه می‌شود. حلقه باید وقتی مقدار صفر می‌شود متوقف شود. 
    با این حال، هیچ ضمانتی وجود ندارد که مقدار <code>item</code> دقیقاً ۰ شود، چون محاسبهٔ ممیز شناور تقریبی است.
  </li>
  <p>
    در عمل ممکن است حلقه بی‌نهایت بار تکرار شود!
  </p>
</div>

```python
item = 1.0
while item > 0:
    item -= 0.1
    print(item)
```

  <h2>معرفی حلقه for</h2>
  <p>
    حلقه <strong>for</strong> در پایتون برای اجرای یک یا چند دستور به تعداد مشخص یا روی یک مجموعه از داده‌ها استفاده می‌شود.
    این مجموعه می‌تواند یک محدوده از اعداد، یک لیست، یک رشته یا هر ساختار قابل تکرار دیگری باشد.
  </p>

  <p>
    با حلقه for می‌توانیم کارهای تکراری را به‌صورت خودکار و منظم انجام دهیم، بدون اینکه نیاز باشد کد را چندین بار کپی کنیم.
  </p>

  <p>
    ساختار کلی حلقه for در پایتون:
  </p>

  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
for variable in sequence:
    # دستورات مورد نظر
    statement(s)
  </pre>

  <h3>مثال ۱: لیست کارهای روزانه</h3>
  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
tasks = ["Check emails", "Attend meeting", "Write report", "Go for a walk", "Read a book"]
for task in tasks:
    print(f"Today's task: {task}")
  </pre>

  <h3>تابع range در پایتون</h3>
  <p>
    تابع <code>range()</code> یکی از پرکاربردترین ابزارها برای ایجاد دنباله‌ای از اعداد در حلقه‌های <strong>for</strong> است.
    این تابع خودش لیست برنمی‌گرداند، بلکه یک شیء قابل تکرار (iterable) تولید می‌کند که می‌توان روی آن پیمایش کرد.
  </p>
  <p>شکل‌های استفاده از <code>range</code>:</p>
  <ul>
    <li><code>range(stop)</code> → تولید اعداد از ۰ تا قبل از <code>stop</code></li>
    <li><code>range(start, stop)</code> → تولید اعداد از <code>start</code> تا قبل از <code>stop</code></li>
    <li><code>range(start, stop, step)</code> → تولید اعداد از <code>start</code> تا قبل از <code>stop</code> با گام <code>step</code></li>
  </ul>

  <h3>مثال ۲: استفاده از range برای چاپ اعداد زوج بین ۲ تا ۱۰</h3>
  <pre style="background-color: #272822; color: #f8f8f2; padding: 10px; border-radius: 8px; direction: ltr; text-align: left;">
for number in range(2, 11, 2):
    print(f"Even number: {number}")
  </pre>

  <p>
    در این مثال:
    <ul>
      <li><code>range(2, 11, 2)</code> اعداد ۲ تا ۱۰ را با گام ۲ تولید می‌کند.</li>
      <li>متغیر <code>number</code> در هر تکرار یکی از این اعداد را می‌گیرد.</li>
      <li>با استفاده از <code>print</code> هر عدد زوج نمایش داده می‌شود.</li>
    </ul>
  </p>

  <hr class="divider" />
</div>

  <h2> دستور continue و pass </h2>

  <li style="margin-bottom: 20px;">
    استفاده از دستور <strong>continue</strong> در داخل یک حلقه باعث می‌شود سایر دستورات آن تکرار اجرا نشوند و اجرای حلقه مستقیماً به تکرار بعدی برود.

  <li style="margin-bottom: 20px;">
    دستور <strong>pass</strong> در پایتون یک دستور خالی است که هیچ کاری انجام نمی‌دهد و معمولاً به‌عنوان جای‌نگهدار (placeholder) استفاده می‌شود تا ساختار کد کامل بماند.

  <h3>مثال ۱: استفاده از continue</h3>

  <p dir='ltr'>var1:1, var2:2</p>

    else:
        print(f"Processing {i}")
  </pre>

  <p>خروجی:</p>
  <p dir='ltr'>Processing 3</p>
  <p dir='ltr'>Processing 4</p>

  <h3>مثال ۳: ترکیب continue و pass</h3>
  <p>در این مثال، وقتی <code>var2 == 1</code> باشد، با <code>continue</code> به تکرار بعدی می‌رویم، و وقتی <code>var2 == 0</code> باشد، با <code>pass</code> هیچ کاری انجام نمی‌دهیم ولی بقیه کد همان تکرار اجرا می‌شود.</p>

for var1 in range(3):
    for var2 in range(3):
        if var2 == 1:
            continue  # Skip this iteration entirely
        if var2 == 0:
            pass  # Do nothing here, but continue with the rest of the code
        print(f"var1:{var1}, var2:{var2}")
  </pre>

  <li style="margin-bottom: 40px;">
    برنامه ای بنویسید که از کاربر دو مقدار صحیح مثبت گرفته و بزرگترین مقسیوم الیه مشترک آنها را محاسبه نماید.

```python
# Prompt the user to enter two integers
n1 = int(input("Enter first integer: "))
n2 = int(input("Enter second integer: "))

gcd = 1
k = 2

while k <= n1 and k <= n2:
    if n1 % k == 0 and n2 % k == 0:
        gcd = k
    k += 1

print(f"The greatest common divisor for {n1} and {n2} is {gcd}")
```
